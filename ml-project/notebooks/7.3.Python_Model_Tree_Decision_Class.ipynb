### **1. Introducción**

Un árbol de decisión es una estructura que incluye un nodo raíz, ramas y nodos hoja. Cada nodo interno representa una prueba sobre un atributo, cada rama representa el resultado de una prueba y cada nodo hoja contiene una etiqueta de clase. El nodo superior del árbol es el nodo raíz.

</div>

### **2. Intuición del Algoritmo**

- Para cada atributo del conjunto de datos, el algoritmo crea un nodo. El atributo más importante se ubica en el nodo raíz.

- Para evaluar la tarea en cuestión, comenzamos en el nodo raíz y descendemos por el árbol siguiendo el nodo correspondiente que cumple con nuestra condición o decisión.

- Este proceso continúa hasta llegar a un nodo hoja. Éste nodo contiene la predicción o el resultado del árbol de decisión.

</div>

<div style="text-align: justify; line-height: 1.5; font-size: 18px;"> 

### **3. Medidas de Selección de Atributos**

Uno de los principales desafíos es seleccionar la mejor variable para realizar cada división del árbol. Este proceso se conoce como **Selección de Atributos**.

Las dos medidas más utilizadas son:

- **Ganancia de Información (Information Gain)**

- **Índice Gini (Gini Index)**

</div>

---

<div style="text-align: justify; line-height: 1.5; font-size: 18px;"> 

### **3.1. Ganancia de Información (Information Gain)**

La Ganancia de Información mide cuánto disminuye la incertidumbre después de realizar una división, el cual se basa en el concepto de **Entropía**, que mide el nivel de impureza o desorden de un conjunto de datos.

**La entropía se define como:**

$$
H(t)=
-\sum_{k=1}^{K}
p(k|t)\log_2(p(k|t))
$$

donde  $p(k|t)$ es la probabilidad de la clase $k$ en el nodo $t$.


**Interpretación:**

- Entropía baja → nodo más puro.

- Entropía alta → nodo más mezclado.

</div>

<div style="text-align: justify; line-height: 1.5; font-size: 18px;"> 

**La Ganancia de Información se calcula como:**

$$
IG =
H(\text{padre})
-
H(\text{hijos})
$$

donde:

- $H(\text{padre})$ = impureza antes de dividir.

- $H(\text{hijos})$ = impureza después de dividir.


**Interpretación:**

- Si la diferencia es grande:

    - la división mejora mucho la separación de clases.

- Si la diferencia es pequeña:

    - la división no aporta mucha información.


El árbol selecciona la variable que produzca la mayor reducción de impureza. El algoritmo **ID3** utiliza este criterio para construir el árbol.

</div>

---

<div style="text-align: justify; line-height: 1.5; font-size: 18px;"> 

### **3.2. Índice Gini (Gini Index)**

El Índice Gini es otra medida de impureza utilizada principalmente por el algoritmo **CART (Classification and Regression Trees)**.

Se define como:

$$
G(t)=
1-
\sum_{k=1}^{K}p(k|t)^2
$$


**Interpretación:**

- $G(t)=0$ → nodo completamente puro.
- Valores altos → mayor mezcla de clases.


**Funcionamiento**

El algoritmo evalúa posibles divisiones y selecciona aquella con el menor Índice Gini.

CART utiliza:

- divisiones binarias,

- variables categóricas o continuas,

- y selecciona el split con menor impureza.

</div>
