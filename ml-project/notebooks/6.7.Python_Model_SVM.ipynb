<h1 align="center"><b>SUPPORT VECTOR REGRESSION (SVR)</b></h1> 

---

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">

La **Regresión de Vectores de Soporte** (SVR) es un algoritmo de aprendizaje supervisado que se utiliza para predecir valores continuos mediante la búsqueda de un hiperplano que se ajuste a los datos dentro de una tolerancia de error definida (tubo insensible). A diferencia de la regresión estándar, la SVR se centra en minimizar el error para los puntos fuera de este tubo, lo que la hace robusta frente a valores atípicos y eficaz tanto para datos lineales como no lineales.

</div>

## **Hiperparámetros del SVR**

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">   

`svm__kernel` : Define el tipo de función que transforma los datos para capturar relaciones.  
- `linear`: relación lineal  
- `rbf`: no lineal (el más usado)  
- `poly`: relaciones polinómicas  

---

`svm__C` : Parámetro de regularización. Controla el equilibrio entre ajustar bien los datos y evitar sobreajuste.  
- Valores altos → modelo más flexible (riesgo de overfitting)  
- Valores bajos → modelo más simple (más regularizado)  

---

`svm__epsilon` : Margen de tolerancia al error. Define qué errores se consideran aceptables sin penalización.  
- Valores altos → más tolerancia (modelo más robusto al ruido)  
- Valores bajos → mayor precisión (pero más sensible)  

---

`svm__gamma` : Define el alcance de influencia de cada punto de datos (solo para kernels no lineales).  
- `scale`: valor automático basado en los datos (recomendado)  
- `auto`: usa una escala más simple  

---

`svm__degree` : Grado del polinomio (solo para kernel `poly`).  
- Valores mayores → modelo más complejo  
- Puede capturar relaciones más sofisticadas, pero aumenta riesgo de sobreajuste  

</div>

---

## **Funcionamiento**

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">  

SVR construye una función que:

- Ajusta los datos dentro de un margen $\epsilon$

- Ignora pequeños errores (ruido)

- Se apoya en unos pocos puntos clave llamados vectores de soporte

  </div>

---

## **Casos de Uso**

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">  

SVR es especialmente útil cuando:

- Hay pocos o medianos datos

- La relación es no lineal

- Se quiere controlar el sobreajuste cuidadosamente

- El ruido en los datos no debe dominar el modelo

  </div>

---

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">   

##  **Definición Formal**

El Support Vector Regression (SVR) es un método de regresión que busca encontrar una función $f(x)$ que se desvíe a lo sumo en un valor $\epsilon$ de las observaciones reales, y que al mismo tiempo sea lo más "plana" posible.

---

### **Formulación**

Dado un conjunto de datos:

$$\{(x_i, y_i)\}_{i=1}^{n}$$

SVR busca una función de la forma:

$$f(x) = w^T \phi(x) + b$$

donde:

- $\phi(x)$: transformación (posiblemente no lineal) al espacio de características  

- $w$: vector de pesos  

- $b$: término de sesgo  

---

 </div>

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">  

## Problema de Optimización

$$\min_{w,b,\xi_i,\xi_i^*} \quad \frac{1}{2} \|w\|^2 + C \sum_{i=1}^{n} (\xi_i + \xi_i^*)$$

sujeto a:

$$
\begin{cases}
y_i - w^T \phi(x_i) - b \leq \epsilon + \xi_i \\
w^T \phi(x_i) + b - y_i \leq \epsilon + \xi_i^* \\
\xi_i, \xi_i^* \geq 0
\end{cases}
$$

donde:

- $\|w\|^2$: controla la **suavidad** del modelo (regularización)  

- $\epsilon$: define el **margen de tolerancia** (epsilon-insensitive tube)  

- $\xi_i, \xi_i^*$: variables de holgura que permiten errores fuera del margen  

- $C$: parámetro que controla el **trade-off entre error y complejidad**  

  </div>

---

---

## **Ejemplo Numérico (SVR)**

Consideremos el siguiente conjunto de datos:

| $x$ | $y$ |
|---|---|
| 1 | 2 |
| 2 | 3 |
| 3 | 2.5 |
| 4 | 4 |

---

### **1. Supongamos un modelo lineal**

$$
f(x) = w x + b
$$

Tomemos una función tentativa:

$$
f(x) = 0.8x + 1
$$

---

### **2. Predicciones**

| $x$ | $y\_real$ | $f(x)$ | $Error$ |
|---|--------|----------|----------------|
| 1 | 2.0 | 1.8 | 0.2 |
| 2 | 3.0 | 2.6 | 0.4 |
| 3 | 2.5 | 3.4 | 0.9 |
| 4 | 4.0 | 4.2 | 0.2 |

---

### **3. Definir margen $\epsilon$**

Supongamos: $$\epsilon = 0.5$$

- Errores menores a 0.5 **no se penalizan**

---

### **4. Identificar puntos dentro y fuera del margen**


| x | Error | ¿Penaliza? |
|---|-------|-----------|
| 1 | 0.2 | No |
| 2 | 0.4 | No |
| 3 | 0.9 | Sí |
| 4 | 0.2 | No |


---

## 5. Variables de Holgura

Para el punto que excede el margen:

$$\xi = 0.9 - 0.5 = 0.4$$

---

## 6. Función Objetivo

SVR minimiza:

$$\frac{1}{2} ||w||^2 + C \sum (\xi_i + \xi_i^*)$$

Supongamos:

- $w = 0.8$ → $\frac{1}{2}w^2 = 0.32$

- $C = 1$  

- Solo hay un error fuera del margen: $\xi = 0.4$  

Entonces:

$$
\text{Costo} = 0.32 + 0.4 = 0.72
$$

---

### **7. Interpretación final**

- Solo el punto $$x=3$$ afecta el modelo  

- Los demás puntos son ignorados (dentro del margen)  

- Ese punto es un **vector de soporte**

- SVR no intenta ajustar todos los puntos $\ldots$ solo los que realmente rompen el margen.

```python
# ====================================== #
# CÓDIGO PARA EL EJEMPLO NUMÉRICO LINEAL #
# ====================================== #

import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4])
y = np.array([2.0, 3.0, 2.5, 4.0])


# Modelo 
w = 0.8
b = 1
epsilon = 0.5

# Función
def f(x):
    return w * x + b

# Valores continuos para graficar
x_vals = np.linspace(0, 5, 100)
y_pred = f(x_vals)

# Clasificación de puntos
errors = np.abs(y - f(X))

inside = errors <= epsilon
outside = errors > epsilon

# Gráfico
plt.figure(figsize=(8,6))

# Recta
plt.plot(x_vals, y_pred, label="f(x) = 0.8x + 1")

# Tubo epsilon
plt.fill_between(
    x_vals,
    y_pred - epsilon,
    y_pred + epsilon,
    alpha=0.2,
    label="Tubo ε"
)

# Límites del tubo
plt.plot(x_vals, y_pred + epsilon, linestyle='--')
plt.plot(x_vals, y_pred - epsilon, linestyle='--')

# Puntos dentro
plt.scatter(X[inside], y[inside], label="Dentro del tubo")

# Puntos fuera (vectores de soporte)
plt.scatter(X[outside], y[outside], label="Vectores de soporte")

# Líneas de error (slack)
for xi, yi in zip(X[outside], y[outside]):
    plt.vlines(xi, f(xi), yi, linestyles='dotted')

plt.title("SVR - Ejemplo Numérico")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()

plt.show()
```

```python
# ========================================= #
# CÓDIGO PARA EL EJEMPLO NUMÉRICO NO LINEAL #
# ========================================= #

import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVR

# Datos simulados
np.random.seed(42)
X = np.sort(5 * np.random.rand(50, 1), axis=0)
y = np.sin(X).ravel()

# Ruido
y += 0.3 * np.random.randn(len(y))

# Modelo SVR
epsilon = 0.3
model = SVR(kernel='rbf', C=10, epsilon=epsilon)
model.fit(X, y)

# Predicciones
y_pred = model.predict(X)

# Identificar puntos
residuals = np.abs(y - y_pred)

inside = residuals <= epsilon
outside = residuals > epsilon

# Gráfico
plt.figure()

# Curva del modelo
X_plot = np.linspace(0, 5, 200).reshape(-1, 1)
y_plot = model.predict(X_plot)
plt.plot(X_plot, y_plot, label="Modelo SVR")

# Tubo epsilon
plt.plot(X_plot, y_plot + epsilon, linestyle="--")
plt.plot(X_plot, y_plot - epsilon, linestyle="--")

# Puntos dentro del tubo
plt.scatter(X[inside], y[inside], label="Dentro del tubo")

# Puntos fuera del tubo
plt.scatter(X[outside], y[outside], label="Fuera del tubo", marker="x")

# Vectores de soporte
plt.scatter(
    model.support_vectors_,
    y[model.support_],
    s=100,
    facecolors='none',
    edgecolors='black',
    label="Vectores de soporte"
)

# Labels
plt.title("SVR con tubo ε")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()

plt.show()
```

1️⃣ **Línea central** : Es la función $f(x)$ aprendida.

2️⃣ **Líneas punteadas** : Son el tubo $\epsilon$, zona donde no hay penalización.

3️⃣ **Puntos dentro** : Error pequeño, no influyen en el modelo.

4️⃣ **Puntos fuera (las X)** : Error grande, sí generan penalización.

5️⃣ **Círculos grandes (vectores de soporte)** : Son los únicos que “mandan” en el modelo.

```python
# ============ #
# 1. LIBRERÍAS #
# ============ #

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from sklearn.svm import SVR

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.inspection import permutation_importance

import warnings
warnings.filterwarnings("ignore")
```

```python
# =================== #
# 2. LECTURA DE RUTAS #
# =================== #

mainpath= "/Users/duvancatano/Documents/Data_Analytics_UdeA/ml-project/data/housing"
filename= "housing_dataset.csv"
fullpath= os.path.join(mainpath,filename)
```

```python
# ================= #
# 3. CARGA DE DATOS #
# ================= #

data = pd.read_csv(fullpath, sep=",") # El separador es "," porque en el archivo .csv los valores están separados por coma
pd.set_option('display.max_columns', None) # Para mostrar todas las columnas del DataFrame sin truncar
```

```python
# ========================= #
# 4. SELECCIÓN DE VARIABLES #
# ========================= #

df = data[[
 'municipio',
 'zona_urbana_rural',
 'tipo_de_vivienda',
 'area_total_m2',
 'habitaciones',
 'banos',
 'antiguedad_de_la_vivienda',
 'parqueaderos',
 'estrato',
 'distancia_al_centro_de_la_ciudad_km',
 'acceso_a_transporte_publico',
 'acceso_a_internet',
 'acceso_a_agua_potable',
 'acceso_a_gas',
 'ingreso_del_hogar',
 'valor_de_mercado_de_la_vivienda'
 ]]
```

```python
# ============================== #
# 5. IDENTIFICACIÓN DE VARIABLES #
# ============================== #

cat_cols = [
    'municipio',
    'zona_urbana_rural',
    'tipo_de_vivienda',
    'acceso_a_transporte_publico',
    'acceso_a_internet',
    'acceso_a_agua_potable',
    'acceso_a_gas'
]

num_cols = [
    'area_total_m2',  
    'ingreso_del_hogar',
    'distancia_al_centro_de_la_ciudad_km'
]

int_cols = [
    'habitaciones', 
    'banos',
    'antiguedad_de_la_vivienda', 
    'parqueaderos',
    'estrato'
]

target = 'valor_de_mercado_de_la_vivienda'
```

```python
# ======================================================================= #
# 6. ONE-HOT ENCODING EN CATEGÓRICAS Y ESTANDARIZACIÓN PARA LAS NUMÉRICAS #
# ======================================================================= #

preprocessor = ColumnTransformer([
    ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), cat_cols),
    ('num', StandardScaler(), num_cols + int_cols)
])
```

```python
# ====================================================== #
# 7. MODELO EN PIPELINE: SUPPORT VECTOR REGRESSION (SVR) #
# ====================================================== #

model = Pipeline([
    ('prep', preprocessor),
    ('svm', SVR())
])
```

```python
# ========================= #
# 8. PARTICIÓN TRAIN / TEST #
# ========================= #

X = df[cat_cols + num_cols + int_cols] 
y = abs(df[target])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
# ============================== #
# 9. BÚSQUEDA DE HIPERPARÁMETROS #
# ============================== #

param_grid = {
    'svm__kernel': ['linear', 'rbf', 'poly'],
    'svm__C': [0.1, 1, 10, 50],
    'svm__epsilon': [0.1, 0.5, 1],
    'svm__gamma': ['scale', 'auto'],
    'svm__degree': [2, 3]  # solo aplica para poly
}

grid = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    scoring='neg_root_mean_squared_error',
    cv=5,
    n_jobs=-1,
    verbose=1
)

grid.fit(X_train, y_train)

print("Mejores parámetros:")
print(grid.best_params_)
```

```python
# ========================= #
# 10. EVALUACIÓN DEL MODELO #
# ========================= #

best_model = grid.best_estimator_

y_train_pred = best_model.predict(X_train)
y_test_pred  = best_model.predict(X_test)

def metrics(y_true, y_pred):
    return {
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "R2": r2_score(y_true, y_pred)
    }

print("TRAIN:", metrics(y_train, y_train_pred))
print("TEST :", metrics(y_test, y_test_pred))
```

```python
# ====================== #
# 11. VALIDACIÓN CRUZADA #
# ====================== #

from sklearn.model_selection import cross_validate

scoring = {
    'rmse': 'neg_root_mean_squared_error',
    'mae' : 'neg_mean_absolute_error',
    'r2'  : 'r2'
}

cv_results = cross_validate(
    best_model,
    X_train,
    y_train,
    scoring=scoring,
    cv=5,
    n_jobs=-1,
    return_train_score=True
)

# Resultados promedio
rmse_cv = -cv_results['test_rmse'].mean()
mae_cv  = -cv_results['test_mae'].mean()
r2_cv   =  cv_results['test_r2'].mean()

print("-------------------")
print("VALIDACIÓN CRUZADA")
print("-------------------")
print(f"RMSE CV: {rmse_cv:.4f}")
print(f"MAE  CV: {mae_cv:.4f}")
print(f"R2   CV: {r2_cv:.4f}")
```

```python
# ===================================== #
# DESVIACIÓN ESTÁNDAR POR MÉTRICA EN CV #
# ===================================== #

print("--------------------")
print("Desviación estándar")
print("--------------------")
print(f"RMSE std: {cv_results['test_rmse'].std():.4f}")
print(f"MAE  std: {cv_results['test_mae'].std():.4f}")
print(f"R2   std: {cv_results['test_r2'].std():.4f}")
```

```python
# ===================== #
# GRÁFICO DE DENSIDADES #
# ===================== #

from scipy.stats import gaussian_kde

# Convertir a positivo
rmse_scores = -cv_results['test_rmse']
mae_scores  = -cv_results['test_mae']
r2_scores   =  cv_results['test_r2']

# Figura
fig, axes = plt.subplots(1, 3, figsize=(15, 4))


# Funciones de densidad
def plot_density(ax, data, title):
    density = gaussian_kde(data)
    xs = np.linspace(min(data), max(data), 100)
    
    ax.plot(xs, density(xs))
    ax.set_title(title)
    ax.set_xlabel("Valor")
    ax.set_ylabel("Densidad")

# Graficar cada métrica
plot_density(axes[0], rmse_scores, "RMSE (CV)")
plot_density(axes[1], mae_scores,  "MAE (CV)")
plot_density(axes[2], r2_scores,   "R2 (CV)")

plt.suptitle("Distribución de métricas - Validación Cruzada", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# ====================== #
# 11. DIAGNÓSTICO VISUAL #
# ====================== #

plt.figure(figsize=(6,4))
plt.scatter(y_test, y_test_pred)
plt.xlabel("Real")
plt.ylabel("Predicho")
plt.title("Predicción vs Real")
plt.show()
```

```python
# ============================================== #
# 12. IMPORTANCIA DE LAS VARIABLES (PERMUTACIÓN) #
# ============================================== #

perm = permutation_importance(
    best_model,
    X_test,
    y_test,
    scoring='neg_root_mean_squared_error',
    n_repeats=10,
    random_state=42,
    n_jobs=-1
)

importance = pd.DataFrame({
    'feature': X.columns,
    'importance': perm.importances_mean
}).sort_values(by='importance', ascending=False)

importance.head(10)
```

```python
# ============================================== #
# 13. GRÁFICO DE LA IMPORTANCIA DE LAS VARIABLES #
# ============================================== #

top = importance.head(10)

plt.figure(figsize=(8,5))
plt.barh(top['feature'][::-1], top['importance'][::-1])
plt.title("Importancia de variables (Permutation)")
plt.show()
```

## **ANÁLISIS DE UNDERFITTING Y OVERFITTING**

```python
# ============================================ #
# ANÁLISIS DE SOBREAJUSTE (TRAIN vs CV vs TEST)
# ============================================ #

from sklearn.model_selection import cross_validate

scoring = {
    'rmse': 'neg_root_mean_squared_error',
    'mae' : 'neg_mean_absolute_error',
    'r2'  : 'r2'
}

y_train_pred = best_model.predict(X_train)

rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
mae_train  = mean_absolute_error(y_train, y_train_pred)
r2_train   = r2_score(y_train, y_train_pred)

cv_results = cross_validate(
    best_model,
    X_train,
    y_train,
    scoring=scoring,
    cv=5,
    n_jobs=-1,
    return_train_score=True
)

rmse_cv = -cv_results['test_rmse'].mean()
mae_cv  = -cv_results['test_mae'].mean()
r2_cv   =  cv_results['test_r2'].mean()

y_test_pred = best_model.predict(X_test)

rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))
mae_test  = mean_absolute_error(y_test, y_test_pred)
r2_test   = r2_score(y_test, y_test_pred)

df_results = pd.DataFrame({
    'Dataset': ['Train', 'CV', 'Test'],
    'RMSE': [rmse_train, rmse_cv, rmse_test],
    'MAE':  [mae_train,  mae_cv,  mae_test],
    'R2':   [r2_train,   r2_cv,   r2_test]
})

df_results
```

```python
# ============================== #
# 14. MODELO CON TODOS LOS DATOS #
# ============================== #

final_model = grid.best_estimator_

final_model.fit(X, y)

import joblib
joblib.dump(final_model, "/Users/duvancatano/Documents/Data_Analytics_UdeA/ml-project/models/model_svm.joblib")
joblib.dump(X.columns.tolist(), "/Users/duvancatano/Documents/Data_Analytics_UdeA/ml-project/models/features_svm.joblib")

print("Modelo SVM guardado")
```
