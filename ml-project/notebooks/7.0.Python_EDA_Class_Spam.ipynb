### **Descripción General de los Datos**

La base de datos corresponde a un conjunto de correos electrónicos utilizado para el análisis y detección automática de spam mediante técnicas de Machine Learning.Cada observación representa un correo electrónico y las variables describen características textuales y tipográficas extraídas de su contenido. Las variables incluyen principalmente:

- Frecuencias de palabras específicas asociadas a correos spam,
- Frecuencias de caracteres especiales,
- Patrones de uso de mayúsculas,
- Términos relacionados con promociones, negocios o tecnología.

Muchas variables representan el porcentaje de aparición de ciertas palabras dentro del correo electrónico, por ejemplo:

- `free`
- `money`
- `credit`
- `business`
- `remove`

las cuales suelen ser frecuentes en mensajes spam.

Además, el conjunto de datos contiene variables relacionadas con:

- Signos de puntuación (`!`, `$`, `#`),
- Cadenas numéricas frecuentes,
- Estadísticas sobre letras mayúsculas.

La variable objetivo es:

<div align="center">

| Variable | Descripción |
|---|---|
| `type` | 1 = Spam, 0= No Spam |

</div>

El objetivo principal es construir modelos de clasificación capaces de identificar automáticamente correos spam a partir de patrones lingüísticos y estadísticos presentes en los mensajes.

````python
# ============ #
# 1. LIBRERÍAS #
# ============ #

import pandas as pd
import numpy as np
import os
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
````

````python
# =================== #
# 2. LECTURA DE RUTAS #
# =================== #

mainpath= "/Users/duvancatano/Documents/Data_Analytics_UdeA/ml-project/data/spam"
filename= "spam_dataset.csv"
fullpath= os.path.join(mainpath,filename)
````

````python
# ================= #
# 3. CARGA DE DATOS #
# ================= #

data = pd.read_csv(fullpath, sep=",") # El separador es "," porque en el archivo .csv los valores están separados por coma
pd.set_option('display.max_columns', None) # Para mostrar todas las columnas del DataFrame sin truncar
data.sample(20)
````

````python
# ====================================================== #
# 4. CODIFICACIÓN DE LA VARIABLE OBJETIVO Y DISTRIBUCIÓN #
# ====================================================== #

data["type"] = np.where(data["type"] == "spam", 1, 0)

# Conteos y porcentajes
conteo = data["type"].value_counts().sort_index()

porcentaje = (
    data["type"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)

# DataFrame resumen
resumen = pd.DataFrame({
    "Conteo": conteo,
    "Porcentaje (%)": porcentaje
})

# Renombrar solo los índices existentes
mapa = {
    0: "No Spam (0)",
    1: "Spam (1)"
}

resumen.index = resumen.index.map(mapa)

# Mostrar resultados

print("Distribución de clases")
print(resumen)
````

## **Funciones Peronalizadas - Valores Perdidos y Nulos**

````python
# Porcentaje de valores perdidos por variable #
def resum_missing(df):
    total = df.isnull().sum().sort_values(ascending=False)
    percent = (df.isnull().sum() *100/df.isnull().count()).sort_values(ascending=False)
    missing_data = pd.concat([total, percent], axis=1, keys=['Total', 'Porcentaje'])
    return missing_data

# Resumen de posición de valores perdidos #
def posmissing(df):
    lista_miss = np.where(df.isna())
    v1 = pd.Series(np.ndarray.tolist(lista_miss[0]))
    v2 = pd.Series(np.ndarray.tolist(lista_miss[1]))
    resumen_vna = pd.concat([v1, v2], axis=1, keys=['posicion_fila', 'posicion_columna'])
    return resumen_vna

# Construcción función que cuenta ceros #
def count_zeros(df):
    total = (df == 0).astype(int).sum(axis=0)
    percent = ((df == 0).astype(int).sum(axis=0) *100/df.count())
    missing_data = pd.concat([total, percent], axis=1, keys=['Total_ceros', 'Porcentaje'])
    return missing_data.sort_values(by='Porcentaje', ascending=False)
````

````python
# ============================== #
# 5. VALORES PERDIDOS Y ATÍPICOS #
# ============================== #

resum_missing(data)
````

````python
# ====================== #
# 6. INFORMACIÓN GENERAL #
# ====================== #

print("="*31)
print("INFORMACIÓN GENERAL DEL DATASET")
print("="*31)

print("\nDimensiones:")
print(data.shape)

print("\nTipos de datos:")
print(data.dtypes)

print("\nValores faltantes:")
print(data.isnull().sum())
````

````python
# =============== #
# 7. DESCRIPTIVOS #
# =============== #

print("\n" + "="*25)
print("ESTADÍSTICAS DESCRIPTIVAS")
print("="*25)

data.describe().T
````

````python
# ==================== #
# 8. BALANCE DE CLASES #
# ==================== #

conteo = data['type'].value_counts().sort_index()

porcentaje = (
    data['type']
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)

resumen = pd.DataFrame({
    'Conteo': conteo,
    'Porcentaje (%)': porcentaje
})

resumen.index = ['No Spam (0)', 'Spam (1)']

print("\nDistribución de clases")
print(resumen)
````

````python
# =============================== #
# 9. HISTOGRAMA VARIABLE OBJETIVO #
# =============================== #

fig = px.bar(
    x=['No Spam (0)', 'Spam (1)'],
    y=conteo.values,
    text=porcentaje.astype(str) + '%',
    color=['No Spam', 'Spam'],
    title='Distribución de la Variable Objetivo'
)

fig.update_layout(
    template='plotly_dark',
    title_x=0.5,
    showlegend=False,
    height=500,
    xaxis_title='Clase',
    yaxis_title='Número de Correos'
)

fig.update_traces(textposition='outside')

fig.show()
````

````python
# =============================== #
# 10. ALGUNAS VARIABLES NUMÉRICAS #
# =============================== #

num_vars = ['num857', 'receive', 'charExclamation', 'business', 'free', 'charDollar', 'remove']
````

````python
# =============== #
# 11. HISTOGRAMAS #
# =============== #

for var in num_vars:

    fig = px.histogram(
        data,
        x=var,
        color='type',
        marginal='box',
        nbins=40,
        opacity=0.7,
        title=f'Distribución de {var}',
        
        # Colores personalizados
        color_discrete_map={
            0: '#636EFA',  # azul -> No Spam
            1: '#EF553B'   # rojo -> Spam
        }
    )

    fig.update_layout(
        template='plotly_dark',
        title_x=0.5,
        height=500,
        legend_title='Spam'
    )

    fig.show()
````

````python
# ============ #
# 12. BOXPLOTS #
# ============ #

for var in num_vars:

    fig = px.box(
        data,
        x='type',
        y=var,
        color='type',
        points='outliers',
        title=f'Boxplot de {var} por clase',

        # Colores personalizados
        color_discrete_map={
            0: '#636EFA',  # azul -> No Spam
            1: '#EF553B'   # rojo -> Spam
        }
    )

    fig.update_layout(
        template='plotly_dark',
        title_x=0.5,
        height=500,
        xaxis_title='Clase',
        yaxis_title=var,
        showlegend=False
    )

    fig.show()
````

````python
# ========================= #
# 13. MATRIZ DE CORRELACIÓN #
# ========================= #

corr = data[num_vars].corr()
fig = px.imshow(
    corr,
    text_auto='.2f',
    aspect='auto',
    color_continuous_scale='RdBu_r',
    title='Matriz de Correlación'
)

fig.update_layout(
    template='plotly_dark',
    title_x=0.5,
    height=900,
    width=1000
)

fig.show()
````

````python
# =========================== #
# 14. DIAGRAMAS DE DISPERSIÓN #
# =========================== #

scatter_pairs = [
    ('capitalTotal', 'capitalLong'),
    ('charExclamation', 'charDollar'),
    ('money', 'free'),
    ('your', 'credit'),
    ('business', 'money')
]

for x_var, y_var in scatter_pairs:

    fig = px.scatter(
        data,
        x=x_var,
        y=y_var,
        color='type',
        opacity=0.7,
        title=f'{x_var} vs {y_var}'
    )

    fig.update_layout(
        template='plotly_dark',
        title_x=0.5,
        height=550
    )

    fig.show()
````

````python
# ================= #
# 15. VIOLIN PLOTS #
# ================= #

important_vars = [
    'free',
    'money',
    'your',
    'credit',
    'charExclamation',
    'capitalTotal'
]

for var in important_vars:

    fig = px.violin(
        data,
        x='type',
        y=var,
        color='type',
        box=True,
        points='all',
        title=f'Violin Plot de {var}'
    )

    fig.update_layout(
        template='plotly_dark',
        title_x=0.5,
        height=500,
        showlegend=False
    )

    fig.show()
````

````python
# ========================= #
# 16. DETECCIÓN DE OUTLIERS #
# ========================= #

Q1 = data[num_vars].quantile(0.25)
Q3 = data[num_vars].quantile(0.75)
IQR = Q3 - Q1

outliers = ((data[num_vars] < (Q1 - 1.5 * IQR)) |
            (data[num_vars] > (Q3 + 1.5 * IQR))).sum()

outliers_df = pd.DataFrame({
    'Variable': outliers.index,
    'Número Outliers': outliers.values
})

fig = px.bar(
    outliers_df.sort_values('Número Outliers', ascending=False),
    x='Variable',
    y='Número Outliers',
    color='Número Outliers',
    title='Cantidad de Outliers por Variable'
)

fig.update_layout(
    template='plotly_dark',
    title_x=0.5,
    height=600,
    xaxis_tickangle=-45
)

fig.show()
````

````python
# ===================== #
# 17. PAIRPLOT REDUCIDO #
# ===================== #

vars_pair = [
    'free',
    'money',
    'credit',
    'charExclamation',
    'capitalTotal',
    'type'
]

fig = px.scatter_matrix(
    data,
    dimensions=vars_pair[:-1],
    color='type',
    title='Scatter Matrix de Variables Relevantes'
)

fig.update_layout(
    template='plotly_dark',
    title_x=0.5,
    height=900,
    width=900
)

fig.show()
````

````python
# =========================== #
# 18. BOXPLOT GLOBAL ESCALADO #
# =========================== #

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled = scaler.fit_transform(data[num_vars])

scaled_df = pd.DataFrame(scaled, columns=num_vars)

scaled_long = scaled_df.melt(
    var_name='Variable',
    value_name='Valor'
)

fig = px.box(
    scaled_long,
    x='Variable',
    y='Valor',
    color='Variable',
    title='Distribución Estandarizada de Variables'
)

fig.update_layout(
    template='plotly_dark',
    title_x=0.5,
    height=700,
    xaxis_tickangle=-60,
    showlegend=False
)

fig.show()
````

````python
# ================================= #
# 19. TOP VARIABLES MÁS IMPORTANTES #
# ================================= #

corr_target = data.corr(numeric_only=True)['type'].drop('type')

importance = (
    corr_target
    .abs()
    .sort_values(ascending=False)
    .head(15)
)

# DataFrame para visualización
importance_df = pd.DataFrame({
    'Variable': importance.index,
    'Importancia': importance.values
})

# Orden ascendente para gráfico horizontal
importance_df = importance_df.sort_values(
    by='Importancia',
    ascending=True
)

fig = px.bar(
    importance_df,
    x='Importancia',
    y='Variable',
    orientation='h',
    color='Importancia',
    text='Importancia',
    color_continuous_scale='Tealgrn',
    title='Top 15 Variables Más Asociadas al Spam',
    template='plotly_dark'
)

fig.update_traces(
    texttemplate='%{text:.3f}',
    textposition='outside',
    marker_line_width=0
)

fig.update_layout(
    title_x=0.5,
    height=700,
    xaxis_title='Correlación Absoluta',
    yaxis_title='',
    coloraxis_showscale=False,
    font=dict(size=14),
    margin=dict(l=20, r=20, t=70, b=20)
)

fig.update_xaxes(
    showgrid=True,
    gridwidth=0.5
)

fig.update_yaxes(
    categoryorder='total ascending'
)

fig.show()
````
