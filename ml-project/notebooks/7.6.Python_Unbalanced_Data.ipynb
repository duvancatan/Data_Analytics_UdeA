<h1 align="center"><b> Clasificación con Datos Desbalanceados </b></h1> 

---

Los datos desbalanceados ocurren cuando una clase tiene muchas más observaciones que otra dentro de un problema de clasificación. Por ejemplo, en detección de fraude puede existir una gran cantidad de transacciones “No Fraude” y pocas transacciones “Fraude”. Este desbalance puede provocar que los modelos de Machine Learning aprendan a favorecer la clase mayoritaria, obteniendo métricas aparentemente altas de **Accuracy** pero con bajo desempeño en la detección de la clase minoritaria. Por esta razón, en problemas desbalanceados es fundamental utilizar métricas como *precision*, *recall*, *F1-score* y *ROC-AUC*, además de técnicas como *class weights*, sobremuestreo (*oversampling*) o submuestreo (*undersampling*) para mejorar el aprendizaje del modelo.

Una de las documentaciones más empleadas para su implementación, puede ser consultada en 🔗 <a href="https://imbalanced-learn.org/stable/index.html"> imbalanced-learn documentation</a>


</div>

### **¿Cómo nos afectan los datos desbalanceados?**

Los datos desbalanceados afectan porque el modelo tiende a aprender más la clase mayoritaria y puede ignorar la clase minoritaria. Como consecuencia:

- puede obtener un *accuracy* alto engañosamente,

- aumentar los *Falsos Negativos*,

- detectar mal casos importantes (spam, fraude, enfermedades),

- generar modelos sesgados hacia la clase dominante.

</div>

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">

## **1. Sobremuestreo de la Clase Minoritaria (Oversamplig)**
Se busca aumentar la representación de la clase minoritaria para equilibrar las distribuciones:

- *Oversampling básico (random)*: repetir muestras aleatorias de la clase minoritaria (aumenta riesgo de sobreajuste).

- *SMOTE y Variantes*: generar nuevas muestras sintéticas interpolando entre ejemplos cercanos de la clase minoritaria:

    - **SMOTE (Synthetic Minority Over-sampling Technique)** : Genera nuevas observaciones sintéticas interpolando entre vecinos cercanos.

    - **SMOTENC/SMOTEN** : SMOTENC mezcla variables numéricas y categóricas, mientras SMOTEN solo categóricas.

    - **bSMOTE** : Genera muestras cerca de la frontera de decisión. Ideal cuando las clases se mezclan.

    - **SVM‑SMOTE** : Usa Support Vector Machines para identificar regiones críticas.

    - **ADASYN** : Genera más muestras en regiones difíciles.

    - **KMeans‑SMOTE** : Agrupa datos con K-Means antes de sintetizar observaciones. Reduce generación en zonas ruidosas.

- *ROSE (Random Over Sampling Examples)*: genera nuevas muestras siguiendo una distribución estimada alrededor de las instancias minoritarias.

Estas técnicas mejoran la capacidad del modelo para “ver” mejor la clase minoritaria, pero pueden aumentar el tiempo de entrenamiento y el riesgo de sobreajuste si se usan en exceso.

</p>

<div align="center">

| Categoría     | Detalles                                                                 |
|---------------|--------------------------------------------------------------------------|
| **Ventajas**  | - Muy simple. <br> - Fácil de implementar. <br> - No pierde información. |
| **Desventajas** | - Incrementa riesgo de sobreajuste. <br> - Duplica ruido y outliers. <br> - No genera nueva información. |
| **Casos de Uso** | - Datasets pequeños. <br> - Modelos simples. <br> - Problemas donde la minoritaria es extremadamente pequeña. |

</div>

<div style="text-align: justify; line-height: 1.5; font-size: 18px;"> 

## **2. Submuestreo de la Clase Mayoritaria (Undersampling)**
El objetivo es reducir observaciones de la clase mayoritaria.

- *Random Under-Sampling* : Elimina aleatoriamente observaciones de la clase mayoritaria.

- *Técnicas Limpiadoras* : Buscan conservar observaciones informativas.

    - **Tomek Links** : Dos observaciones de clases opuestas forman un Tomek Link si son vecinos más cercanos mutuos.

    - **Edited Nearest Neighbours (ENN)** : Elimina observaciones mal clasificadas por sus vecinos.

    - **NearMiss** : Selecciona ejemplos mayoritarios cercanos a la minoritaria.

    - **One-Sided Selection (OSS)** : Combina Tomek Links y Condensed Nearest Neighbour.

    - **Neighbourhood Cleaning Rule (NCL)** : Usa ENN para limpiar observaciones conflictivas.

- *Cluster Centroids* : Representa la clase mayoritaria mediante centroides.

<div align="center">

| Categoría       | Detalles                                                                 |
|----------------|--------------------------------------------------------------------------|
| **Ventajas**    | - Reduce el desbalance y el tiempo de entrenamiento. <br> - Disminuye costo computacional y memoria. <br> - Útil en datasets grandes y fácil de implementar. |
| **Desventajas** | - Riesgo de perder información y patrones relevantes. <br> - Puede generar inestabilidad y menor capacidad predictiva. <br> - No crea nueva información y es sensible a la selección de datos. |

</div>

## **3. Combinaciones de Sobremuestreo y Submuestreo**
Buscan equilibrar y limpiar simultáneamente.

- *SMOTE + Tomek Links* : Aplicar SMOTE. Eliminar Tomek Links.

- *SMOTE + ENN* : Generar muestras sintéticas. Limpiar ruido usando ENN.

## **4. Métodos Basados en Ensambles**
Integran balanceo dentro del algoritmo.

- **Easy Ensemble** : Entrena múltiples modelos usando distintos subconjuntos balanceados.

- **Balanced Random Forest** : Cada árbol se entrena sobre muestras balanceadas.

Para cada árbol:

$$D_b = D_{minoritaria} \cup D_{majoritaria}^{sample}$$
	​
- **Balanced Bagging** : Cada clasificador base recibe un dataset balanceado.

- **RUSBoost** : Combina Random Under-Sampling y AdaBoost

| **Desventajas** | - Mayor costo computacional, más complejo y lento que Random Forest estándar. <br> - Sensible al tamaño de muestra, al ruido y puede sobreajustar. <br> - Mayor consumo computacional y complejidad. |

## **Nota:**
No existe un único método óptimo para todos los problemas de desbalance de clases. La elección puede depender de:

- Tamaño del dataset.
- Nivel de ruido.
- Complejidad de la frontera.
- Tipo de variables.
- Costo computacional.
- Algoritmo de clasificación utilizado.
- Probar múltiples estrategias.
- Evaluar con métricas adecuadas:
    - Recall
    - Precision
    - F1-score
    - ROC-AUC
    - PR-AUC
    - Validar mediante cross-validation estratificada.

````python
# ================= #
# Cargar Librearías #
# ================= #


import warnings
import pandas as pd
import numpy as np
import os
import joblib
from pathlib import Path

# Visualización
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go


# Preprocesamiento
from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Métricas
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score
)

# Modelos
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# Técnicas de balanceo
from imblearn.over_sampling import (
    RandomOverSampler,
    SMOTE,
    ADASYN
)

from imblearn.under_sampling import (
    RandomUnderSampler,
    TomekLinks,
    EditedNearestNeighbours
)

from imblearn.combine import (
    SMOTETomek,
    SMOTEENN
)

from imblearn.ensemble import (
    BalancedRandomForestClassifier,
    EasyEnsembleClassifier,
    RUSBoostClassifier,
    BalancedBaggingClassifier
)

# Pipeline de imblearn
from imblearn.pipeline import Pipeline as ImbPipeline
````

````python
# ============================= #
# Selección de Varibales Insumo #
# ============================= #

var = [
 'FRAUDE',
 'VALOR',
 'COD_PAIS',
 'CANAL',
 'DIASEM',
 'SEXO',
 'SEGMENTO',
 'EDAD',
 'INGRESOS',
 'EGRESOS',
 'NROPAISES',
 'NROCIUDADES',
 'Dist_Sum_INTER',
 'Dist_Mean_INTER',
 'Dist_Max_INTER',
 'Dist_max_NAL',
 'Dist_Mean_NAL',
 'Dist_HOY',
 'Dist_sum_NAL',
 'QUINCENA',
 'PERIODO_DIA']
df = data[var]
````

````python
# ========= #
# VARIABLES # 
# ========= #

target = "FRAUDE"

X = df.drop(columns=target)
y = df[target]
````

````python
# ================================= #
# VARIABLES NUMÉRICAS Y CATEGÓRICAS #
# ================================= #

numeric_features = [
    "VALOR",            
    "EDAD",             
    "INGRESOS",         
    "EGRESOS",          
    "NROPAISES",        
    "NROCIUDADES",      
    "Dist_Sum_INTER",   
    "Dist_Mean_INTER",  
    "Dist_Max_INTER",   
    "Dist_max_NAL",     
    "Dist_Mean_NAL",    
    "Dist_HOY",         
    "Dist_sum_NAL" 
]

categorical_features = [
     "COD_PAIS",     
     "CANAL",        
     "DIASEM",       
     "SEXO",         
     "SEGMENTO",     
     "QUINCENA",     
     "PERIODO_DIA"
]
````

````python
# ================================= #
# PREPROCESAMIENTO CON ESCALAMIENTO #
# ================================= #

numeric_transformer_scaled = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor_scaled = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer_scaled, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
````

````python
# ================================= #
# PREPROCESAMIENTO SIN ESCALAMIENTO #
# ================================= #

numeric_transformer_no_scaled = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

preprocessor_no_scaled = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer_no_scaled, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
````

````python
# ================ #
# TRAIN TEST SPLIT #
# ================ #

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=123
)
````

````python
# ===================== #
# FUNCIÓN DE EVALUACIÓN #
# ===================== #

def evaluate_model(model, model_name):
    print(model_name)


    # ENTRENAMIENTO
    model.fit(X_train, y_train)

    # PREDICCIONES
    y_pred = model.predict(X_test)

    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:,1]
    else:
        y_prob = None

    # CLASSIFICATION REPORT
    print("\nClassification Report\n")
    print(classification_report(y_test, y_pred))

    # MATRIZ DE CONFUSIÓN
    cm = confusion_matrix(y_test, y_pred)
    vp = cm[1,1]
    fp = cm[0,1]
    fn = cm[1,0]
    vn = cm[0,0]

    cm_custom = pd.DataFrame(
        [
            [vp, fp],
            [fn, vn]
        ],
        index=[
            "Predicho Positivo",
            "Predicho Negativo"
        ],
        columns=[
            "Real Positivo",
            "Real Negativo"
        ]
    )

    # LABELS PERSONALIZADOS
    labels = [
        [
            f"VP<br>{vp}",
            f"FP<br>{fp}"
        ],
        [
            f"FN<br>{fn}",
            f"VN<br>{vn}"
        ]
    ]

    # MATRIZ DE CONFUSIÓN - PLOTLY
    fig = px.imshow(
        cm_custom,
        text_auto=False,
        color_continuous_scale=[
            [0.0, "#7f0000"],
            [0.5, "#1a1a1a"],
            [1.0, "#006400"]
        ],
        aspect="auto",
        title=f"Matriz de Confusión - {model_name}"
    )

    # Agregar texto manual
    for i in range(2):
        for j in range(2):

            fig.add_annotation(
                x=cm_custom.columns[j],
                y=cm_custom.index[i],
                text=labels[i][j],
                showarrow=False,
                font=dict(
                    size=22,
                    color="white"
                )
            )

    fig.update_layout(
        width=700,
        height=500,
        title_x=0.5,
        template="plotly_dark",
        coloraxis_showscale=False
    )

    fig.update_xaxes(
        title="Clases reales",
        side="top"
    )

    fig.update_yaxes(
        title="Clases predichas"
    )

    fig.show()

    # MÉTRICAS
    metrics = {
        "Model": model_name,
        "Accuracy": round((y_pred == y_test).mean(), 4),
        "Precision": round(
            precision_score(y_test, y_pred), 4
        ),
        "Recall": round(
            recall_score(y_test, y_pred), 4
        ),
        "F1": round(
            f1_score(y_test, y_pred), 4
        )
    }

    if y_prob is not None:

        metrics["ROC_AUC"] = round(
            roc_auc_score(y_test, y_prob), 4
        )

    # DATAFRAME MÉTRICAS
    metrics_df = pd.DataFrame([metrics])
    print("\nMÉTRICAS\n")
    display(
    metrics_df.style.hide(axis="index")
)
````

````python
# ================================= #
# MODELO BASE - REGRESIÓN LOGÍSTICA #
# ================================= #

baseline_model = ImbPipeline(
    steps=[
        ("preprocessor", preprocessor_scaled),
        (
            "model",
            LogisticRegression(
                max_iter=500,
                class_weight="balanced"
            )
        )
    ]
)

evaluate_model(
    baseline_model,
    "Modelo Logístico"
)
````

## **Balanced Random Forest Classifier**

---

### **Idea principal**

En un Random Forest tradicional, donde cada árbol se entrena con muestras bootstrap del dataset original.


### **¿Qué hace BalancedRandomForest?**

La idea es balancear las clases para CADA árbol antes de entrenarlo.

Proceso interno

| Clase       | Cantidad |
|-------------|----------|
| No fraude   | 10,000   |
| Fraude      | 500      |

---

- ### **1. Bootstrap de la clase minoritaria**

El algoritmo toma:

TODAS o muchas observaciones minoritarias

por ejemplo:

500 fraudes

---


- ### **2. Under-sampling de la mayoritaria**

Luego selecciona aleatoriamente:

500 no-fraudes

---

- ### **3. Dataset balanceado**

Cada árbol recibe:

| Clase       | Cantidad |
|-------------|----------|
| No fraude   | 500      |
| Fraude      | 500      |

---

- ### **4. Entrenar árbol**

Ahora el árbol aprende sobre datos balanceados.

</div>

---

````python
# ====================== #
# BALANCED RANDOM FOREST #
# ====================== #

from imblearn.ensemble import BalancedRandomForestClassifier
from imblearn.pipeline import Pipeline as ImbPipeline
````

````python
# ====== #
# MODELO #
# ====== #

brf_model = ImbPipeline(
    steps=[

        # ----------------------------------------------------
        # PREPROCESAMIENTO
        # ----------------------------------------------------

        ("preprocessor", preprocessor_no_scaled),

        # ----------------------------------------------------
        # MODELO
        # ----------------------------------------------------

        (
            "model",

            BalancedRandomForestClassifier(

                n_estimators=300,
                random_state=123,
                n_jobs=-1,

                # Opcional
                replacement=True,
                sampling_strategy="all"
            )
        )
    ]
)
````

````python
# ============= #
# ENTRENAMIENTO #
# ============= #

brf_model.fit(X_train, y_train)
````

````python
# ============ #
# PREDICCIONES #
# ============ #

y_pred = brf_model.predict(X_test)
y_prob = brf_model.predict_proba(X_test)[:,1]
````
