La <b>regresión isotónica</b> es un método no paramétrico de estimación que permite modelar la relación entre una variable explicativa y una variable respuesta bajo la restricción de <i>monotonía</i>, es decir, asumiendo que dicha relación es consistentemente creciente o decreciente. A diferencia de modelos paramétricos, no se impone una forma funcional específica, lo que le permite adaptarse de manera flexible a la estructura de los datos. Este enfoque es especialmente útil cuando se tiene conocimiento previo sobre el comportamiento ordenado de las variables, como en la calibración de probabilidades o en procesos donde se espera una relación estructural. El ajuste se obtiene minimizando una función de pérdida cuadrática ponderada, garantizando simultáneamente que los valores estimados respeten la restricción de orden, lo cual conduce a una función estimada típicamente escalonada.

$$
\min_{\hat{y}_1, \dots, \hat{y}_n} \sum_{i=1}^{n} w_i \,(y_i - \hat{y}_i)^2
\quad \text{sujeto a} \quad
\hat{y}_1 \leq \hat{y}_2 \leq \dots \leq \hat{y}_n.
$$

<p align="center">
   <img src="img/isoto.png" width="700" height="450">
</p>

En esta expresión, $y_i$ representa los valores observados, $\hat{y}_i$ los valores ajustados, y $w_i > 0$ son pesos asociados a cada observación, los cuales permiten asignar distinta importancia a los datos. La función objetivo minimiza la suma de los errores cuadráticos ponderados, mientras que la restricción asegura que los valores estimados sigan una estructura monótonamente creciente (o decreciente, si se invierte la desigualdad), garantizando coherencia con el comportamiento esperado de la relación entre variables.

---

<h3>- Características principales</h3>

<ul>
<li>No asume una forma funcional específica.</li>
<li>Garantiza una relación monótona entre variables.</li>
<li>Se basa en el algoritmo <i>Pool Adjacent Violators Algorithm (PAVA)</i>.</li>
<li>Produce una función escalonada (piecewise constant).</li>
</ul>

---

<h3>- Interpretación</h3>

<p>
La Regresión Isotónica ajusta los valores de tal forma que se respeten las restricciones de orden, corrigiendo posibles violaciones de monotonía en los datos originales. Esto la hace especialmente útil en contextos donde se espera una relación estructural entre variables.
</p>

---

<h3>- Casos de uso</h3>

<ul>
<li>Calibración de probabilidades en modelos de clasificación.</li>
<li>Modelos donde se conoce una relación creciente o decreciente.</li>
<li>Post-procesamiento de predicciones para imponer coherencia.</li>
</ul>

</div>

```python
# ============ #
# 1. LIBRERÍAS #
# ============ #

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

from matplotlib.collections import LineCollection
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split, cross_validate
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.isotonic import IsotonicRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.metrics import mean_absolute_percentage_error
from sklearn.utils import check_random_state
```

```python
# =================== #
# 2. LECTURA DE RUTAS #
# =================== #

mainpath= "/Users/duvancatano/Documents/Data_Analytics_UdeA/ml-project/data/housing"
filename= "housing_dataset.csv"
fullpath= os.path.join(mainpath,filename)
```

```python
# ================= #
# 3. CARGA DE DATOS #
# ================= #

data = pd.read_csv(fullpath, sep=",") # El separador es "," porque en el archivo .csv los valores están separados por coma
pd.set_option('display.max_columns', None) # Para mostrar todas las columnas del DataFrame sin truncar
```

```python
# ========================= #
# 4. SELECCIÓN DE VARIABLES #
# ========================= #

X_var = 'area_total_m2'
y_var = 'valor_de_mercado_de_la_vivienda'

data = data[[X_var, y_var]].dropna()
```

```python
# ================ #
# 5. ORDENAR DATOS #
# ================ #

data = data.sort_values(by=X_var)

X = data[X_var].values
y = data[y_var].values
```

```python
# ================================ #
# 6. AJUSTE DE REGRESIÓN ISOTÓNICA #
# ================================ #

iso = IsotonicRegression(increasing=True)

y_iso = iso.fit_transform(X, y)
```

```python
# ============================= #
# 7. AJUSTE DE REGRESIÓN LINEAL #
# ============================= #

lr = LinearRegression()

X_reshape = X.reshape(-1, 1)
lr.fit(X_reshape, y)

y_lr = lr.predict(X_reshape)
```

```python
# ================ #
# 8. VISUALIZACIÓN #
# ================ #

plt.figure(figsize=(10,6))

plt.scatter(X, y, alpha=0.3, label='Datos reales')
plt.plot(X, y_iso, color='red', label='Isotónica')
plt.plot(X, y_lr, linestyle='--', label='Lineal')

plt.xlabel(X_var)
plt.ylabel(y_var)
plt.title('Regresión Isotónica vs Lineal')

plt.legend()
plt.show()
```

```python
# ======================== #
# 9. EVALUACIÓN DE MODELOS #
# ======================== #

def metrics(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    return mae, rmse, r2

mae_iso, rmse_iso, r2_iso = metrics(y, y_iso)
mae_lr, rmse_lr, r2_lr = metrics(y, y_lr)

print("🔵 Isotónica")
print("MAE:", mae_iso)
print("RMSE:", rmse_iso)
print("R2:", r2_iso)

print("\n🔴 Lineal")
print("MAE:", mae_lr)
print("RMSE:", rmse_lr)
print("R2:", r2_lr)
```
