<h1 align="center"><b>RANDOM FOREST - REGRESIÓN</b></h1> 

---

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">

El Random Forest es una extensión de los árboles de decisión que busca mejorar su desempeño mediante la combinación de múltiples árboles construidos de manera aleatoria. Mientras un árbol individual puede ser altamente sensible a los datos (lo que puede llevar a sobreajuste), Random Forest reduce esta variabilidad entrenando varios árboles sobre distintas muestras del conjunto de datos y promediando sus predicciones.

En el contexto de regresión, cada árbol genera una predicción numérica y el modelo final corresponde al promedio de todas las predicciones individuales, logrando así un estimador más estable y con mejor capacidad de generalización. Este enfoque se basa en dos ideas fundamentales: el muestreo aleatorio de datos (bootstrap) y la selección aleatoria de variables en cada división, lo que introduce diversidad entre los árboles.

Adicionalmente, conserva la capacidad de los árboles para capturar relaciones no lineales, pero corrige su principal debilidad —*el sobreajuste*— convirtiéndose en uno de los modelos más robustos y utilizados en analítica de datos y aprendizaje automático.
