<div style="text-align: justify; line-height: 1.5; font-size: 18px;">

Un modelo Random Forest está compuesto por un conjunto (**ensemble**) de Árboles de Decisión individuales. Cada uno de estos árboles es entrenado con una muestra aleatoria con reemplazo, extraída de los datos de entrenamiento originales mediante (*Bootstrapping*). Esto implica que cada árbol se entrena con un conjunto de datos ligeramente diferente. 

En cada árbol individual, las observaciones se distribuyen a través de bifurcaciones (nodos), dando forma a la estructura del árbol hasta llegar a un nodo terminal. La predicción de una nueva observación se obtiene al agregar las predicciones de todos los árboles individuales que conforman el modelo.

Para comprender cómo funcionan los modelos **Random Forest** es fundamental familiarizarse primero con los conceptos de **Ensemble**, **Bagging** y **Boosting**.

<h1 align="center"><b>Ensemble - Bagging - Boosting</b></h1> 

---

<div style="text-align: justify; line-height: 1.5; font-size: 18px; display: grid; grid-template-columns: repeat(3, 1fr); gap: 20px;">

<div>

***Ensemble:*** En regresión, es una estrategia que combina las predicciones de múltiples modelos para obtener un resultado más preciso y estable que el de un solo modelo. La idea central es que distintos modelos pueden capturar diferentes aspectos de los datos, y al integrarlos se reduce el error global y se mejora la capacidad de generalización.

</div>

<div>

***Bagging:*** El Bagging (Bootstrap Aggregating) consiste en entrenar varios modelos de forma independiente utilizando distintas muestras aleatorias del conjunto de datos (con reemplazo) y luego promediar sus predicciones. Este enfoque reduce la *variabilidad* del modelo y lo hace más robusto frente a cambios en los datos, siendo especialmente útil en modelos inestables como los árboles de decisión.

</div>

<div>

**Boosting:** El Boosting es una técnica en la que los modelos se construyen de manera secuencial, donde cada nuevo modelo se enfoca en corregir los errores cometidos por los anteriores. A través de este proceso iterativo, el modelo final logra capturar patrones más complejos y mejorar progresivamente su desempeño, reduciendo principalmente el *sesgo*.

</div>

</div>

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">

- **Varianza (Variance)** : Es la sensibilidad del modelo a pequeñas variaciones en los datos de entrenamiento. Una alta varianza implica que el modelo cambia mucho entre muestras y tiende a sobreajustar (overfitting).

- **Sesgo (Bias)**: Es el error sistemático que comete un modelo por ser demasiado simple para capturar la relación real entre las variables. Un alto sesgo implica subajuste (underfitting).

</div>

## **Definición :**


<div style="text-align: justify; line-height: 1.5; font-size: 18px;">    

El **Random Forest** es un método de aprendizaje supervisado basado en *ensambles*, que construye un conjunto de árboles de decisión $$\{T_b\}_{b=1}^{B},$$ a partir de muestras aleatorias del conjunto de datos y combina sus predicciones para obtener un estimador más robusto.

En el caso de regresión, el modelo se define como:

$$
\hat{f}(x) = \frac{1}{B} \sum_{b=1}^{B} T_b(x)
$$

donde:

- $B$ es el número total de árboles en el bosque,  

- $T_b(x)$ es la predicción del árbol $b$-ésimo, construido sobre una muestra *bootstrap* (muestreo aleatoria con reemplazo) del conjunto de datos.

- En cada nodo de cada árbol se selecciona aleatoriamente un subconjunto de variables para realizar la partición.

---

**Observaciones**

- Este procedimiento introduce aleatoriedad tanto en los datos como en las variables, lo que reduce la correlación entre los árboles y, en consecuencia, disminuye la varianza del modelo final sin incrementar significativamente el sesgo.

- Esta metodología puede ser interpretada como el promedio de múltiples árboles de decisión entrenados sobre versiones aleatorias de los datos, lo que permite obtener modelos más estables y con mejor capacidad de generalización.


</div>

## **Out-of-Bag Error**

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">    

Dada la naturaleza del proceso de bagging, resulta posible estimar el error de test sin necesidad de recurrir a métodos de validación cruzada (cross-validation). 

El hecho de que los árboles se ajusten empleando muestras generadas por bootstrapping conlleva que, en promedio, cada ajuste use solo aproximadamente dos tercios de las observaciones originales. Al tercio restante se le llama out-of-bag (OOB).

Si para cada árbol ajustado en el proceso de bagging se registran las observaciones empleadas, se puede predecir la respuesta de la observación $i$ haciendo uso de aquellos árboles en los que esa observación ha sido excluida y promediándolos (la moda en el caso de los árboles de clasificación). Siguiendo este proceso, se pueden obtener las predicciones para las $n$ observaciones y con ellas calcular el OOB-mean square error (para regresión) o el OOB-classification error (para árboles de clasificación). Como la variable respuesta de cada observación se predice empleando únicamente los árboles en cuyo ajuste no participó dicha observación, el OOB-error sirve como estimación del error de test. De hecho, si el número de árboles es suficientemente alto, el OOB-error es prácticamente equivalente al leave-one-out cross-validation error.

Esta es una ventaja añadida de los métodos de bagging, y por lo tanto de **Random Forest** ya que evita tener que recurrir al proceso de validación cruzada (computacionalmente costoso) para la optimización de los hiperparámetros.

</div>

## **Importancia de los predictores**

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">    

Si bien es cierto que el proceso de bagging (Random Forest) consigue mejorar la capacidad predictiva en comparación a los modelos basados en un único árbol, esto tiene un coste asociado, la interpretabilidad del modelo se reduce. Al tratarse de una combinación de múltiples árboles, no es posible obtener una representación gráfica sencilla del modelo y no es inmediato identificar de forma visual *qué predictores son más importantes*. 

Sin embargo, se han desarrollado nuevas estrategias para cuantificar la importancia de los predictores que hacen de los modelos de bagging (Random Forest) una herramienta muy potente, no solo para predecir, sino también para el análisis exploratorio. Dos de estas medidas son: importancia por permutación e impureza de nodos.

---

### **Importancia por permutación**

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">    

Es una forma directa y agnóstica al modelo de medir qué tan importante es una variable, rompe su información y se observa cuánto empeora el modelo.

Para cada variable $X_j$:

1. Medir el desempeño del modelo ($RMSE$ o $R^2$)
2. Permutas los valores de $X_j$
3. Vuelves a predecir
4. Mides cuánto empeora el desempeño

$$Imp(X_j) = Error_perm(j) - Error_original$$

- Si el error **sube mucho** → la variable era **importante**
- Si **casi no cambia** → la variable **aporta poco**

</div>

---

### **Incremento de la pureza de nodos** (Emplado por ```scikit-learn```)

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">    

Supongamos un nodo con valores de $Y$ muy dispersos. Eso es impuro (alta varianza en regresión). Cuando se hace un split "bueno", se divide ese nodo en dos grupos donde los valores son más parecidos entre sí, es decir, más puros (menos varianza).

El árbol busca maximizar la reducción de impureza en cada split:

$$\Delta I = I_{\text{antes}} - I_{\text{despues}}$$

Donde:

- $I_{\text{antes}}$ : impureza del nodo padre  
- $I_{\text{despues}}$ : promedio ponderado de los hijos  

- Si $\Delta I$ es **grande** → el split fue **muy bueno**
- Si $\Delta I$ es **pequeño** → el split **aporta poco**

Intuitivamente, una variable es importante si ayuda a separar bien los datos en cada paso del árbol.

#### **Importancia en el Bosque**

Se promedia sobre todos los árboles:

$$Imp_j = \frac{1}{T} \sum_{t=1}^{T} Imp_j^{(t)},$$

el cual da como resultado **Importancia Global**, donde $$Imp_j^{(\text{árbol})} = \sum (\text{reducciones de impureza donde se usó } X_j).$$

</div>

## **Hiperparámetros**

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">  </div>

---

 **Hiperparámetros para tuning en Random Forest**

- ```n_estimators``` : Número de árboles. Aumentarlo mejora estabilidad, pero con retornos decrecientes. Empieza con 100–300.

- ```max_depth``` : Profundidad máxima del árbol. Controla el sobreajuste. Limitarlo suele mejorar generalización.
 
- ```min_samples_split``` : Mínimo de observaciones para dividir un nodo. Valores mayores → árboles más simples.
 
- ```min_samples_leaf``` : Mínimo de observaciones en hojas. Muy útil para evitar ruido y mejorar estabilidad.
 
- ```max_features``` : Número de variables consideradas en cada split. Clave para reducir correlación entre árboles. Usar ```sqrt``` o ```log2``` es buena práctica.

---

**Hiperparámetros de ajuste fino**

- ```bootstrap``` : Indica si se usa muestreo con reemplazo. Generalmente se deja en True.

- ```oob_score``` : Permite validar el modelo sin test set. Muy útil en exploración.

- ```max_samples``` : Número de muestras por árbol. Útil en datasets grandes para acelerar entrenamiento.

- ```min_impurity_decrease``` : Umbral mínimo para dividir. Ayuda a controlar splits innecesarios.

- ```max_leaf_nodes``` : Límite de hojas. Alternativa a controlar profundidad.

---

**Hiperparámetros avanzados**

- ```criterion``` : Métrica de división. Normalmente no se cambia (```squared_error```).

- ```min_weight_fraction_leaf``` : Relevante solo si trabajas con pesos en observaciones.

- ```ccp_alpha``` : Parámetro de poda. Más común en árboles individuales que en Random Forest.

- ```monotonic_cst``` : Restricciones de monotonicidad. Uso avanzado en contextos específicos.

---

**Configuración técnica**

- ```n_jobs``` : Paralelización. Usar -1 en práctica.

- ```random_state``` : Reproducibilidad.

- ```verbose``` : Nivel de mensajes en consola.

- ```warm_start``` : Permite seguir entrenando el modelo agregando más árboles.

````python
# ============ #
# 1. LIBRERÍAS #
# ============ #

import pandas as pd
import numpy as np
import os

from sklearn.model_selection import train_test_split, cross_validate
from sklearn.ensemble import RandomForestRegressor
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
````

````python
# =================== #
# 2. LECTURA DE RUTAS #
# =================== #

mainpath= "/Users/duvancatano/Documents/Data_Analytics_UdeA/ml-project/data/housing"
filename= "housing_dataset.csv"
fullpath= os.path.join(mainpath,filename)
````

````python
# ================= #
# 3. CARGA DE DATOS #
# ================= #

data = pd.read_csv(fullpath, sep=",") # El separador es "," porque en el archivo .csv los valores están separados por coma
pd.set_option('display.max_columns', None) # Para mostrar todas las columnas del DataFrame sin truncar
````

````python
# ========================= #
# 4. SELECCIÓN DE VARIABLES #
# ========================= #

df = data[[
 'municipio',
 'zona_urbana_rural',
 'tipo_de_vivienda',
 'area_total_m2',
 'habitaciones',
 'banos',
 'antiguedad_de_la_vivienda',
 'parqueaderos',
 'estrato',
 'distancia_al_centro_de_la_ciudad_km',
 'acceso_a_transporte_publico',
 'acceso_a_internet',
 'acceso_a_agua_potable',
 'acceso_a_gas',
 'ingreso_del_hogar',
 'valor_de_mercado_de_la_vivienda'
 ]]
````

````python
# ============================== #
# 5. IDENTIFICACIÓN DE VARIABLES #
# ============================== #

cat_cols = [
    'municipio',
    'zona_urbana_rural',
    'tipo_de_vivienda',
    'acceso_a_transporte_publico',
    'acceso_a_internet',
    'acceso_a_agua_potable',
    'acceso_a_gas'
]

num_cols = [
    'area_total_m2',  
    'ingreso_del_hogar',
    'distancia_al_centro_de_la_ciudad_km'
]

int_cols = [
    'habitaciones', 
    'banos',
    'antiguedad_de_la_vivienda', 
    'parqueaderos',
    'estrato'
]

target = 'valor_de_mercado_de_la_vivienda'
````

````python
# ================================================================================================================== #
# 6. ONE-HOT ENCODING (A las variables categóricas aplícales One-Hot Encoding, y a las numéricas déjalas como están) #
# ================================================================================================================== #

preprocessor = ColumnTransformer([
    ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), cat_cols),
    ('num', 'passthrough', num_cols + int_cols)
])
````

````python
# ========================= #
# 7. MODELO : RANDOM FOREST #
# ========================= #

rf = RandomForestRegressor(
    n_estimators=200,     # El número de árboles en el bosque
    max_depth=None,       # La profundidad máxima del árbol. Si es None, los nodos se expanden hasta que todas las hojas sean puras 
    min_samples_split=5,  # El número mínimo de muestras necesarias para dividir un nodo interno
    min_samples_leaf=2,   # El número mínimo de muestras requerido para un nodo hoja
    oob_score=True,       # Indica si se deben usar muestras fuera de la bolsa para estimar la puntuación de generalización. Por defecto, usa r2_score 
    random_state=42,      # Controla tanto la aleatoriedad del remuestreo de las muestras utilizadas al construir árboles 
    n_jobs=-1             # Número de trabajos que se ejecutarán en paralelo. -1 significa usar todos los procesadores.
)
````

````python
# =========== #
# 8. PIPELINE #
# =========== #

model = Pipeline([
    ('prep', preprocessor),
    ('rf', rf)
])
````

````python
# ========================= #
# 9. PARTICIÓN TRAIN / TEST #
# ========================= #

X = df[cat_cols + num_cols + int_cols] 
y = abs(df[target])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
````

````python
# ====================== #
# 7. ENTRENADO EL MODELO #
# ====================== #

model.fit(X_train, y_train)
````

````python
# =================== #
# 8. OUT-OF-BAG ERROR #
# =================== #

oob_score = model.named_steps['rf'].oob_score_
print("OOB Score (R2):", oob_score)
````

````python
# ===================== #
# 9. EVALUACIÓN EN TEST #
# ===================== #

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

mape = np.mean(np.abs((y_test - y_pred) / y_test))

def smape(y_true, y_pred):
    return np.mean(
        np.abs(y_true - y_pred) / ((np.abs(y_true) + np.abs(y_pred)) / 2)
    )

print("MÉTRICAS EN TEST")
print("MAE:", mae)
print("RMSE:", rmse)
print("R2:", r2)
print("MAPE:", mape)
print("SMAPE:", smape(y_test, y_pred))
````

````python
# ====================== #
# 10. VALIDACIÓN CRUZADA #
# ====================== #

scoring = {
    'mae': 'neg_mean_absolute_error',
    'rmse': 'neg_root_mean_squared_error',
    'r2': 'r2'
}

cv_results = cross_validate(model, X, y, cv=10, scoring=scoring)

print("CV MAE:", -cv_results['test_mae'].mean())
print("CV RMSE:", -cv_results['test_rmse'].mean())
print("CV R2:", cv_results['test_r2'].mean())
````

````python
# ================================ #
# 11. IMPORTANCIA DE LAS VARIABLES #
# ================================ #

# NOTA: Aquí estamos evaluando el modelo inicial con hiperparámetros dados a priori

rf_model = model.named_steps['rf']
feature_names = model.named_steps['prep'].get_feature_names_out()

importance = pd.Series(rf_model.feature_importances_, index=feature_names)
importance = importance.sort_values(ascending=False)

importance.head(20)
````

````python
# ======================================= #
# GRÁFICO DE IMPORTANCIA DE LAS VARIABLES #
# ======================================= #

import matplotlib.pyplot as plt

importance.head(10).plot(kind='barh')
plt.gca().invert_yaxis()
plt.title("Top 10 Variables más importantes")
plt.show()
````

Cabe tener en cuenta que, cuando se busca el valor óptimo de un hiperparámetro con dos métricas distintas, el resultado obtenido raramente es el mismo. Lo importante es que ambas métricas identifiquen las mismas regiones de interés.

````python
# ======================================== #
# LIBRERÍA NECESARIAS PARA LA OPTIMIZACIÓN #
# ======================================== #

# Datos
import numpy as np
import pandas as pd

# Gráficos
import matplotlib.pyplot as plt
plt.style.use('fivethirtyeight')
plt.rcParams['lines.linewidth'] = 1.5
plt.rcParams['font.size'] = 8

# Preprocesado y modelado
import sklearn
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
from sklearn.model_selection import RepeatedKFold
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import ParameterGrid
from sklearn.model_selection import cross_val_score
from sklearn.inspection import permutation_importance
from joblib import Parallel, delayed, cpu_count
import optuna

# Configuración warnings
import warnings
optuna.logging.set_verbosity(optuna.logging.WARNING)

print(f"Versión de scikit-learn: {sklearn.__version__}")
````

---

### **Número de árboles** (```n_estimators ```)

<div style="text-align: justify; line-height: 1.5; font-size: 18px;"> 

En Random Forest, el número de árboles no es un hiperparámetro crítico en cuanto que, añadir árboles, solo puede hacer que mejorar el resultado. En Random Forest no se produce overfitting por exceso de árboles. Sin embargo, añadir árboles una vez que la mejora se estabiliza es una perdida te recursos computacionales.

</div>

---

````python
# ======================================== #
# VALIDACIÓN EMPLEANDO EL OUT-OF-BAG ERROR #
# ======================================== #

import matplotlib.pyplot as plt
import numpy as np
import warnings

warnings.filterwarnings('ignore')

train_scores = []
oob_scores   = []

estimator_range = range(10, 201, 10) 

for n in estimator_range:
    
    rf = RandomForestRegressor(
        n_estimators=n,
        oob_score=True,
        random_state=42,
        n_jobs=-1
    )
    
    model = Pipeline([
        ('prep', preprocessor),
        ('rf', rf)
    ])
    
    model.fit(X_train, y_train)
    
    train_scores.append(model.score(X_train, y_train))
    oob_scores.append(model.named_steps['rf'].oob_score_)

# ================== #
# IDENTIFICAR ÓPTIMO #
# ================== #

best_idx = np.argmax(oob_scores)
best_n   = estimator_range[best_idx]
best_oob = oob_scores[best_idx]

# ======== #
# GRÁFICO  #
# ======== #

plt.figure(figsize=(12, 7))

plt.plot(estimator_range, train_scores, marker='o', label="Train R²")
plt.plot(estimator_range, oob_scores, marker='s', label="OOB R²")

# Punto óptimo
plt.scatter(best_n, best_oob, s=120, label="Mejor modelo")

# Línea vertical en el óptimo
plt.axvline(best_n, linestyle='--', alpha=0.7)

# Texto del óptimo
plt.text(best_n, best_oob,
         f" n={best_n}\n R²={best_oob:.3f}",
         verticalalignment='bottom')

plt.title("Evolución del OOB Score vs Número de Árboles", fontsize=16)
plt.xlabel("Número de Árboles (n_estimators)", fontsize=12)
plt.ylabel("R²", fontsize=12)

plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

plt.tight_layout()
plt.show()

# ========= #
# RESULTADO #
# ========= #

print(f"Mejor n_estimators: {best_n}")
print(f"Mejor OOB R²: {best_oob:.4f}")

warnings.filterwarnings('default')
````

````python
# ============================================================== #
# Detectar cuándo el modelo deja de aprender significativamente  #
# y entra en zona plana. Esto es decisión automática, no visual. #
# ============================================================== #

def find_plateau(x, y, tol=1e-3, window=3):
    """
    Detecta el primer punto donde la mejora es marginal.
    """
    diffs = np.diff(y)
    
    for i in range(len(diffs) - window):
        if np.all(diffs[i:i+window] < tol):
            return x[i]
    
    return x[-1]


optimal_n = find_plateau(list(estimator_range), oob_scores)

print(f"n_estimators óptimo (criterio saturación): {optimal_n}")
````

````python
# ================================= #
# Interpretación                    #
# Si OOB ≈  CV →  Confiable         #
# Si OOB >> CV →  Optimista         #
# Si OOB << CV →  Modelo Inestable  #
# ================================= # 

from sklearn.model_selection import cross_val_score

rf = RandomForestRegressor(
    n_estimators=optimal_n,
    random_state=42,
    n_jobs=-1
)

model = Pipeline([
    ('prep', preprocessor),
    ('rf', rf)
])

cv_scores = cross_val_score(model, X, y, cv=5, scoring='r2')

print("OOB R²:", oob_scores[estimator_range.index(optimal_n)])
print("CV  R²:", cv_scores.mean())
````

El valor de ```max_features``` es uno de los hiperparámetros más importantes de random forest, ya que controla cuántas variables (features) puede considerar el modelo en cada split de un árbol.



En scikit-learn puede tomar valores:

- ```float``` : proporción (ej: 0.5 = 50\%)

- ```None``` : usa todas las variables

````python
# ===================== #
# OOB vs max_features 2 #
# ===================== #

import numpy as np
import matplotlib.pyplot as plt

train_scores = []
oob_scores   = []

# Usar número de features después del preprocessing (esto evita inconsistencias con OneHotEncoder)
n_features = model.named_steps['prep'].fit(X_train).transform(X_train).shape[1]

max_features_range = range(1, n_features + 1, max(1, n_features // 20))  

for mf in max_features_range:
    
    rf = RandomForestRegressor(
        n_estimators=100,
        max_features=mf,
        oob_score=True,
        random_state=42,
        n_jobs=-1
    )
    
    model = Pipeline([
        ('prep', preprocessor),
        ('rf', rf)
    ])
    
    model.fit(X_train, y_train)
    
    train_scores.append(model.score(X_train, y_train))
    oob_scores.append(model.named_steps['rf'].oob_score_)

# ====== #
# ÓPTIMO #
# ====== #

best_idx = np.argmax(oob_scores)
best_mf  = list(max_features_range)[best_idx]
best_oob = oob_scores[best_idx]

# ================ #
# GRÁFICO MEJORADO #
# ================ #

plt.figure(figsize=(12, 7))

plt.plot(max_features_range, train_scores, marker='o', label="Train R²")
plt.plot(max_features_range, oob_scores, marker='s', label="OOB R²")

# Punto óptimo
plt.scatter(best_mf, best_oob, s=120, label="Mejor modelo")

# Línea vertical
plt.axvline(best_mf, linestyle='--', alpha=0.7)

# Texto
plt.text(best_mf, best_oob,
         f" max_features={best_mf}\n R²={best_oob:.3f}",
         verticalalignment='bottom')

# Estética
plt.title("OOB Score vs Número de Variables (max_features)", fontsize=16)
plt.xlabel("max_features", fontsize=12)
plt.ylabel("R²", fontsize=12)

plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

plt.tight_layout()
plt.show()

print(f"Mejor max_features: {best_mf}")
print(f"Mejor OOB R²: {best_oob:.4f}")
````

````python
# ==================== #
# CV vs max_features 3 #
# ==================== #

import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import cross_val_score
from sklearn.metrics import mean_squared_error

train_rmse = []
cv_rmse    = []

# =========================== #
# FEATURES REALES (post-prep) #
# =========================== #

prep_temp = Pipeline([('prep', preprocessor)])
X_trans = prep_temp.fit_transform(X_train)

n_features = X_trans.shape[1]

# Menos puntos → gráfico interpretable
step = max(1, n_features // 20)
max_features_range = range(1, n_features + 1, step)

# ==== #
# LOOP #
# ==== #

for mf in max_features_range:
    
    rf = RandomForestRegressor(
        n_estimators=100,
        max_features=mf,
        random_state=42,
        n_jobs=-1
    )
    
    model = Pipeline([
        ('prep', preprocessor),
        ('rf', rf)
    ])
    
    # =========== #
    # TRAIN ERROR
    # =========== #
    model.fit(X_train, y_train)
    y_pred = model.predict(X_train)
    
    rmse_train = np.sqrt(mean_squared_error(y_train, y_pred))
    train_rmse.append(rmse_train)
    
    # ================ #
    # CROSS VALIDATION #
    # ================ #
    scores = cross_val_score(
        model,
        X_train,
        y_train,
        scoring='neg_root_mean_squared_error',
        cv=5,
        n_jobs=-1
    )
    
    cv_rmse.append(-scores.mean())

# ==================== #
# ÓPTIMO (mínimo RMSE) #
# ==================== #

best_idx = np.argmin(cv_rmse)
best_mf  = list(max_features_range)[best_idx]
best_rmse = cv_rmse[best_idx]

# ========== #
# SATURACIÓN #
# ========== #

def find_plateau_min(x, y, tol=1e-3, window=3):
    diffs = -np.diff(y)
    for i in range(len(diffs) - window):
        if np.all(diffs[i:i+window] < tol):
            return x[i]
    return x[np.argmin(y)]

sat_mf = find_plateau_min(list(max_features_range), cv_rmse)
sat_rmse = cv_rmse[list(max_features_range).index(sat_mf)]

# ======= #
# GRÁFICO #
# ======= #

plt.figure(figsize=(12, 7))

plt.plot(max_features_range, train_rmse, marker='o', label="Train RMSE")
plt.plot(max_features_range, cv_rmse, marker='s', label="CV RMSE")

# mínimo
plt.scatter(best_mf, best_rmse, s=120, label="Mínimo RMSE")

# saturación
plt.scatter(sat_mf, sat_rmse, s=120, marker='X', label="Saturación")

plt.axvline(sat_mf, linestyle='--', alpha=0.7)

plt.text(sat_mf, sat_rmse,
         f" sat={sat_mf}\n RMSE={sat_rmse:.2f}",
         verticalalignment='bottom')

plt.title("RMSE vs max_features (con CV)", fontsize=16)
plt.xlabel("max_features", fontsize=12)
plt.ylabel("RMSE", fontsize=12)

plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

plt.tight_layout()
plt.show()

# ========== #
# RESULTADOS #
# ========== #

print(f"Mejor (mínimo): max_features = {best_mf}")
print(f"Saturación: max_features = {sat_mf}")
````

````python
# ========================== #
# GRID SEARCH PARALELO (OOB) #
# ========================== #

import pandas as pd
import numpy as np
from joblib import Parallel, delayed
from multiprocessing import cpu_count
from sklearn.model_selection import ParameterGrid
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error

# ================ #
# GRID INTELIGENTE #
# ================ #

param_grid = ParameterGrid({
    'rf__n_estimators': [100],
    'rf__max_features': ['sqrt', 0.3, 0.5],
    'rf__max_depth': [None, 10, 20],
    'rf__min_samples_leaf': [1, 3, 5]
})

# ===================== #
# FUNCIÓN DE EVALUACIÓN #
# ===================== #

def eval_oob(X, y, params, verbose=False):
    
    model = Pipeline([
        ('prep', preprocessor),
        ('rf', RandomForestRegressor(
            oob_score=True,
            n_jobs=1, 
            random_state=42
        ))
    ])
    
    model.set_params(**params)
    model.fit(X, y)
    
    rf_model = model.named_steps['rf']
    
    # OOB R2
    oob_r2 = rf_model.oob_score_
    
    # OOB RMSE
    y_oob = rf_model.oob_prediction_
    rmse_oob = np.sqrt(mean_squared_error(y, y_oob))
    
    if verbose:
        print(f"Modelo: {params} ✓")
    
    return {
        **params,
        'oob_r2': oob_r2,
        'oob_rmse': rmse_oob
    }

# ============== #
# PARALELIZACIÓN #
# ============== #

resultados = Parallel(n_jobs=cpu_count()-1)(
    delayed(eval_oob)(X_train, y_train, params)
    for params in param_grid
)

# ========== #
# RESULTADOS #
# ========== #

resultados = pd.DataFrame(resultados)

# Ordenar por RMSE 
resultados = resultados.sort_values('oob_rmse')

# Top resultados
resultados.head(10)
````

````python
# ====================== #
# CONSTRUIR MODELO FINAL #
# ====================== #

final_model = Pipeline([
    ('prep', preprocessor),
    ('rf', RandomForestRegressor(
        oob_score=True,
        n_jobs=-1,
        random_state=42
    ))
])

final_model.set_params(**best_params)
````

### **Bayesian Search**

<div style="text-align: justify; line-height: 1.5; font-size: 18px;"> 

La búsqueda **Grid Search** y **Random Search** pueden generar buenos resultados, sobre todo cuando se reduce el rango de exploración. Sin embargo, ninguna de ellas tiene en cuenta los resultados obtenidos hasta el momento, lo que les impide centrar la búsqueda en las regiones de mayor interés y evitar las innecesarias.

Una alternativa es utilizar métodos de optimización bayesiana para buscar hiperparámetros. En términos generales, la optimización bayesiana de hiperparámetros consiste en crear un modelo probabilístico en el que la función objetivo es la métrica de validación del modelo (RMSE, AUC, entre otras). Con esta estrategia, la búsqueda se redirige en cada iteración a las regiones de mayor interés. El objetivo final es reducir el número de combinaciones de hiperparámetros con las que se evalúa el modelo, eligiendo sólo los mejores candidatos. Este enfoque es especialmente ventajoso cuando el espacio de búsqueda es muy grande o la evaluación del modelo es muy lenta.

</div>

---

````python
# ====================== #
# OPTUNA + RANDOM FOREST #
# ====================== #

import optuna
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score, RepeatedKFold

# ============================
# FUNCIÓN OBJETIVO
# ============================

def objective(trial):
    
    params = {
        'rf__n_estimators': trial.suggest_int('rf__n_estimators', 100, 400, step=50),
        'rf__max_depth': trial.suggest_categorical('rf__max_depth', [None, 8, 10, 15, 20]),
        'rf__min_samples_split': trial.suggest_int('rf__min_samples_split', 2, 20),
        'rf__min_samples_leaf': trial.suggest_int('rf__min_samples_leaf', 1, 10),
        'rf__max_features': trial.suggest_categorical('rf__max_features', ['sqrt', 0.3, 0.5]),
        'rf__ccp_alpha': trial.suggest_float('rf__ccp_alpha', 0.0, 0.05)
    }

    model = Pipeline([
        ('prep', preprocessor),
        ('rf', RandomForestRegressor(
            oob_score=False,   # no usar con CV
            n_jobs=1,          # evitar nested parallelism
            random_state=42
        ))
    ])
    
    model.set_params(**params)

    cv = RepeatedKFold(n_splits=5, n_repeats=2, random_state=42)

    scores = cross_val_score(
        estimator=model,
        X=X_train,
        y=y_train,
        cv=cv,
        scoring='neg_root_mean_squared_error',
        n_jobs=-1
    )
    
    return -scores.mean()   # RMSE positivo

# ============ #
# OPTIMIZACIÓN #
# ============ #

study = optuna.create_study(direction='minimize')  # ahora minimizamos RMSE
study.optimize(objective, n_trials=40, timeout=60*5, show_progress_bar=True)

# ========== 
# RESULTADOS
# ========== 

print("Mejores hiperparámetros:")
print(study.best_params)

print("\nMejor RMSE (CV):")
print(study.best_value)
````

````python
# ======================================= #
# MODELO FINAL (OPTUNA) + EVALUACIÓN TEST #
# ======================================= #

from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# 1. Construir modelo con Pipeline
modelo_final = Pipeline([
    ('prep', preprocessor),
    ('rf', RandomForestRegressor(
        oob_score=False,   # no usar con test
        n_jobs=-1,
        random_state=42
    ))
])

# 2. Asignar mejores parámetros
modelo_final.set_params(**study.best_params)

# 3. Entrenar
modelo_final.fit(X_train, y_train)

# 4. Predicciones
y_pred = modelo_final.predict(X_test)

# 5. Métricas completas
mae  = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2   = r2_score(y_test, y_pred)

print(" Resultados en TEST:")
print(f"MAE : {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R2  : {r2:.4f}")
````
