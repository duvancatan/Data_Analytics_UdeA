<h1 align="center"><b> XGBOOST (eXtreme Gradient Boosting) - REGRESIÓN</b></h1> 

---

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">

Es un algoritmo de aprendizaje supervisado basado en **Boosting**, donde los modelos (generalmente árboles de decisión) se construyen de forma secuencial, de modo que cada nuevo árbol intenta corregir los errores cometidos por los anteriores. A diferencia de métodos como **Random Forest**, donde los árboles se entrenan de manera independiente, **XGBoost** aprende de los residuos paso a paso, optimizando una función de pérdida mediante técnicas de gradiente.

Además, incorpora mecanismos de regularización, control de complejidad y optimización computacional, que lo hace altamente eficiente y preciso en problemas de *Regresión* y *Clasificación*, siendo uno de los algoritmos más utilizados en la práctica profesional. Intuitivamente, el *Random Forest* aprende en paralelo, mientras que el *XGBoost* aprende secuencial.

</div>

   <img src="img/xgb.png" width="1200" height="400">
</p>

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">

**1. Aprende de los errores** : Cada árbol se enfoca en lo que el modelo anterior no logró capturar.

**2. Optimización por gradiente** : XGBoost usa gradiente para encontrar las mejores mejoras en cada paso.

**3. Regularización** : Evita sobreajuste usando penalizaciones en la complejidad de los árboles.

**4. Eficiente y escalable** : Diseñado para ser rápido y manejar grandes volúmenes de datos.

</div>

---

### **Ventajas**

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">

- Sirven para regresión y clasificación.

- Manejan predictores numéricos y categóricos (sin necesidad de codificación en teoría).

- No requieren distribuciones específicas ni estandarización.

- Robustos ante valores atípicos.

- Pueden predecir incluso con datos faltantes (aunque con menos precisión).

- Útiles para explorar e identificar variables importantes.

- Escalables para muchos datos.

</div>

---

### **Desventajas**

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">

- Menos interpretables que un solo árbol (muchos árboles combinados).

- Pérdida de información al discretizar predictores continuos.

- Favorecen predictores continuos o con muchos niveles por azar.

- No pueden extrapolar fuera del rango de entrenamiento.

</div>

## **Definición** 

El modelo **XGBoost** (Extreme Gradient Boosting) es un método de aprendizaje supervisado que construye un modelo aditivo de la forma:

$$
\hat{y}_i = \sum_{m=1}^{M} f_m(x_i), \quad f_m \in \mathcal{F},
$$

donde cada $f_m$ es un árbol de decisión y $\mathcal{F}$ es el espacio de árboles posibles. El modelo se entrena minimizando una función objetivo regularizada:

$$\mathcal{L} = \sum_{i=1}^{n} l(y_i, \hat{y}_i) + \sum_{m=1}^{M} \Omega(f_m),$$

en la cual:

---

- **Predicción  $\hat{y}_i$** : Corresponde a la predicción del modelo para la observación $i$, obtenida como la suma de las contribuciones de todos los árboles. Cada árbol aporta una corrección al modelo.

---

- **Función Base  $f_m(x_i)$** : Cada  $f_m$ representa un árbol de decisión. Son modelos *débiles* que se construyen secuencialmente y aprenden a corregir los errores de los árboles anteriores.

---

- **Número de Árboles  $M$** : Indica la cantidad total de árboles en el modelo. Controla la complejidad; más árboles implican mayor capacidad de ajuste, pero también mayor riesgo de *sobreajuste*.

---

- **Función de Pérdida  $l(y_i, \hat{y}_i)$** : Mide el error entre los valores reales y las predicciones del modelo. Por ejemplo:

    a. **Regresión**: Error Cuadrático Medio $(MSE).$

    b. **Clasificación**: $log-loss.$

Es el componente principal que el modelo busca minimizar.

---

- **Regularización  $\Omega(f_m)$** : Penaliza la complejidad de los árboles para evitar sobreajuste:

$$\Omega(f) = \gamma T + \frac{1}{2} \lambda \sum_{j=1}^{T} w_j^2$$

donde:

- $T$ : Número de hojas del árbol  

- $w_j$ : Valor asignado a cada hoja  

Este término controla el tamaño del árbol y la magnitud de sus predicciones.

---

- **Interpretación**

    a. Modelo aditivo → suma de árboles

    b. Aprendizaje secuencial → cada árbol corrige errores

    c. Optimización → usa gradiente y hessiano

    d. Regularización → controla complejidad

</div>

## **Funcionamiento**

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">

Consideremos el siguiente conjunto de datos donde se explica el funcionamiento de XGBoost con gradiente y hessiano explícitos :

<div align="center">

| x | y |
| - | - |
| 1 | 3 |
| 2 | 5 |
| 3 | 7 |
| 4 | 10 |

</div>

En este caso usaremos la siguiente Función de Pérdida:

$$ l(y,\hat{y})=\frac{1}{2}(y-\hat{y})^2 $$

</div>

---

### **1. Predicción Inicial**

$$\hat{y}^{(0)} = \frac{3 + 5 + 7 + 10}{4} = 6.25$$


| x | y  | pred |
| - | -- | ---- |
| 1 | 3  | 6.25 |
| 2 | 5  | 6.25 |
| 3 | 7  | 6.25 |
| 4 | 10 | 6.25 |


---

### **2. Gradiente y Hessiano**

$$g_i = \hat{y} - y, \ \ \ \ h_i=1$$

| x | y  | pred | $g_i$ |
| - | -- | ---- | ----- |
| 1 | 3  | 6.25 | 3.25  |
| 2 | 5  | 6.25 | 1.25  |
| 3 | 7  | 6.25 | -0.75 |
| 4 | 10 | 6.25 | -3.75 |


---

### **3. Construir Árbol - Split Simple**

$$x < 2.5$$

---

<div style="display: flex; gap: 40px;">

<div style="flex: 1;">

**Nodo Izquierdo** ($x = 1 , 2$)

$G_L = 3.25 + 1.25 = 4.5$

$H_L = 2$

</div>

<div style="flex: 1;">

**Nodo Derecho** ($x = 3 , 4$)

$G_R = -0.75 - 3.75 = -4.5$

$H_R = 2$

</div>

</div>

---

### **4. Valor de Hojas**

$$\omega = - \frac{G}{H+\lambda},$$

donde supondremos que $\lambda = 1.$


<div style="display: flex; gap: 40px;">

<div style="flex: 1;">

---

**Izquierdo**

$\omega = - \frac{4.5}{2+1} = -1.5$

</div>

<div style="flex: 1;">

---

**Derecho**

$\omega = - \frac{-4.5}{2+1} = 1.5$

</div>

</div>


<div align="center">



## Árbol 1

 $x<2.5 → -1.5$

 $x≥2.5 → +1.5$

</div>

---

### **5. Actualización**

Con $\eta = 0.3:$

$$\hat{y}^{(1)} = \hat{y}^{(0)} + \eta \cdot \omega$$

````python
# ============ #
# 1. LIBRERÍAS #
# ============ #

import numpy as np
import pandas as pd
import os

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_score, RepeatedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder


from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from xgboost import XGBRegressor # pip install xgboost 

import warnings
warnings.filterwarnings("ignore")
````

````python
# =================== #
# 2. LECTURA DE RUTAS #
# =================== #

mainpath= "/Users/duvancatano/Documents/Data_Analytics_UdeA/ml-project/data/housing"
filename= "housing_dataset.csv"
fullpath= os.path.join(mainpath,filename)
````

````python
# ================= #
# 3. CARGA DE DATOS #
# ================= #

data = pd.read_csv(fullpath, sep=",") # El separador es "," porque en el archivo .csv los valores están separados por coma
pd.set_option('display.max_columns', None) # Para mostrar todas las columnas del DataFrame sin truncar
````

````python
# ========================= #
# 4. SELECCIÓN DE VARIABLES #
# ========================= #

df = data[[
 'municipio',
 'zona_urbana_rural',
 'tipo_de_vivienda',
 'area_total_m2',
 'habitaciones',
 'banos',
 'antiguedad_de_la_vivienda',
 'parqueaderos',
 'estrato',
 'distancia_al_centro_de_la_ciudad_km',
 'acceso_a_transporte_publico',
 'acceso_a_internet',
 'acceso_a_agua_potable',
 'acceso_a_gas',
 'ingreso_del_hogar',
 'valor_de_mercado_de_la_vivienda'
 ]]
````

````python
# ============================== #
# 5. IDENTIFICACIÓN DE VARIABLES #
# ============================== #

cat_cols = [
    'municipio',
    'zona_urbana_rural',
    'tipo_de_vivienda',
    'acceso_a_transporte_publico',
    'acceso_a_internet',
    'acceso_a_agua_potable',
    'acceso_a_gas'
]

num_cols = [
    'area_total_m2',  
    'ingreso_del_hogar',
    'distancia_al_centro_de_la_ciudad_km'
]

int_cols = [
    'habitaciones', 
    'banos',
    'antiguedad_de_la_vivienda', 
    'parqueaderos',
    'estrato'
]

target = 'valor_de_mercado_de_la_vivienda'
````

````python
# ================================================================================================================== #
# 6. ONE-HOT ENCODING (A las variables categóricas aplícales One-Hot Encoding, y a las numéricas déjalas como están) #
# ================================================================================================================== #

preprocessor = ColumnTransformer([
    ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), cat_cols),
    ('num', 'passthrough', num_cols + int_cols)
])
````

````python
# =============== #
# 7. MODELO : XGB #
# =============== #

xgb = XGBRegressor(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1 # Utilizará todos los núcleos disponibles en la CPU para realizar tareas en paralelo.
    #n_jobs=1 # La tarea (entrenamiento, búsqueda de hiperparámetros) utilizará un solo núcleo del CPU.
)       
````

````python
# =========== #
# 8. PIPELINE #
# =========== #

model = Pipeline([
    ('prep', preprocessor),
    ('xgb', xgb)
])
````

````python
# ========================= #
# 9. PARTICIÓN TRAIN / TEST #
# ========================= #

X = df[cat_cols + num_cols + int_cols] 
y = abs(df[target])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
````

````python
# ======================= #
# 10. ENTRENADO EL MODELO #
# ======================= #

model.fit(X_train, y_train)
````

````python
# ====================== #
# 11. EVALUACIÓN EN TEST #
# ====================== #

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

mape = np.mean(np.abs((y_test - y_pred) / y_test))

def smape(y_true, y_pred):
    return np.mean(
        np.abs(y_true - y_pred) / ((np.abs(y_true) + np.abs(y_pred)) / 2)
    )

print("MÉTRICAS EN TEST")
print("MAE:", mae)
print("RMSE:", rmse)
print("R2:", r2)
print("MAPE:", mape)
print("SMAPE:", smape(y_test, y_pred))
````

````python
# ====================== #
# 12. VALIDACIÓN CRUZADA #
# ====================== #

scoring = {
    'mae': 'neg_mean_absolute_error',
    'rmse': 'neg_root_mean_squared_error',
    'r2': 'r2'
}

cv_results = cross_validate(model, X, y, cv=10, scoring=scoring)

print("CV MAE:", -cv_results['test_mae'].mean())
print("CV RMSE:", -cv_results['test_rmse'].mean())
print("CV R2:", cv_results['test_r2'].mean())
````

## **Importancia de variables por reducción de la impureza (Gain)**

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">

La importancia de variables basada en la reducción de la impureza mide cuánto contribuye cada variable a mejorar el modelo, evaluando la disminución del error que produce en los nodos donde es utilizada para realizar particiones.

El procedimiento consiste en:

Construir los árboles del modelo.
Identificar en qué nodos se utiliza cada variable.
Calcular cuánto reduce el error en cada split.
Sumar esas reducciones a lo largo de todos los árboles.

Si una variable genera grandes reducciones de error en los splits, se considera más importante para el modelo.

Este método es rápido de calcular y está integrado en modelos basados en árboles, aunque puede estar sesgado hacia variables que permiten más particiones.

</div>

---

````python
# ================================ #
# 11. IMPORTANCIA DE LAS VARIABLES #
# ================================ #

# NOTA: Aquí estamos evaluando el modelo inicial con hiperparámetros dados a priori

rf_model = model.named_steps['xgb']
feature_names = model.named_steps['prep'].get_feature_names_out()

importance = pd.Series(rf_model.feature_importances_, index=feature_names)
importance = importance.sort_values(ascending=False)

importance.head(20)
````

````python
# ======================================= #
# GRÁFICO DE IMPORTANCIA DE LAS VARIABLES #
# ======================================= #

importance.head(10).plot(kind='barh')
plt.gca().invert_yaxis()
plt.title("Top 10 Variables más importantes")
plt.show()
````

## **Importancia de variables por permutación**

<div style="text-align: justify; line-height: 1.5; font-size: 18px;">

La **importancia por permutación** mide el impacto de cada variable en el desempeño del modelo, evaluando cuánto empeora la métrica de error cuando los valores de dicha variable son desordenados aleatoriamente.

El procedimiento consiste en:

1. Calcular el desempeño original del modelo.
2. Permutar aleatoriamente una variable.
3. Volver a predecir con el modelo.
4. Medir el cambio en el error.

Si al permutar una variable el error aumenta significativamente, se considera que dicha variable es importante para el modelo.

Este método es independiente del modelo y refleja el impacto real de cada variable en la capacidad predictiva.

</div>

---

````python
# =========================== #
# IMPORTANCIA POR PERMUTACIÓN #
# =========================== #

from sklearn.inspection import permutation_importance

perm = permutation_importance(
    model,
    X_train,
    y_train,
    scoring='neg_root_mean_squared_error',
    n_repeats=10,
    random_state=42,
    n_jobs=-1
)

perm_df = pd.DataFrame({
    'feature': X_train.columns,
    'importance': perm.importances_mean
}).sort_values(by='importance', ascending=False)

perm_df.head(10)
````

````python
# ====================== #
# GRÁFICO DE PERMUTACIÓN #
# ====================== #

top_perm = perm_df.head(10)

plt.figure(figsize=(8,5))
plt.barh(top_perm['feature'][::-1], top_perm['importance'][::-1])
plt.title("Top variables (Permutación)")
plt.show()
````

### **Hiperparámetros a Optimizar**

| Hiperparámetro           | Descripción |
|-------------------------|------------|
| `xgb__n_estimators`     | Número de árboles que se construyen secuencialmente. Más árboles aumentan la capacidad del modelo, pero también el riesgo de sobreajuste. |
| `xgb__max_depth`        | Profundidad máxima de cada árbol. Controla la complejidad del modelo; valores altos capturan patrones más complejos pero pueden sobreajustar. |
| `xgb__learning_rate`    | Tasa de aprendizaje que escala la contribución de cada árbol. Valores pequeños hacen el aprendizaje más lento pero más estable. |
| `xgb__subsample`        | Proporción de observaciones utilizadas para entrenar cada árbol. Introduce aleatoriedad y ayuda a reducir el sobreajuste. |
| `xgb__colsample_bytree` | Proporción de variables (features) usadas en cada árbol. Similar a Random Forest, reduce correlación entre árboles. |
| `xgb__gamma`            | Ganancia mínima requerida para realizar una partición en un nodo. Valores altos hacen el modelo más conservador (menos splits). |
| `xgb__reg_alpha`        | Término de regularización L1 (tipo LASSO). Favorece modelos más simples eliminando variables poco relevantes. |
| `xgb__reg_lambda`       | Término de regularización L2 (tipo Ridge). Penaliza valores grandes en los pesos para evitar sobreajuste. |

</div>

````python
# ============================== #
# 1. RANDOMIZED SEARCH - XGBOOST #
# ============================== #

from sklearn.model_selection import RandomizedSearchCV, RepeatedKFold
from scipy.stats import randint, uniform


# Pipeline base
model = Pipeline([
    ('prep', preprocessor),
    ('xgb', XGBRegressor(
        objective='reg:squarederror',
        random_state=42,
        n_jobs=1   # Evitar paralelismo anidado y Cada modelo usa 1 núcleo
    ))
])

# n_jobs=1 (XGB)
# n_jobs=-1 (RandomizedSearch)

# Espacio de búsqueda
param_dist = {
    'xgb__n_estimators': randint(100, 500),
    'xgb__max_depth': randint(3, 10),
    'xgb__learning_rate': uniform(0.01, 0.2),
    'xgb__subsample': uniform(0.6, 0.4),        # [0.6, 1.0]
    'xgb__colsample_bytree': uniform(0.6, 0.4),
    'xgb__gamma': uniform(0, 2),
    'xgb__reg_alpha': uniform(0, 1),
    'xgb__reg_lambda': uniform(0.5, 2)
}

# Validación cruzada
cv = RepeatedKFold(
    n_splits=5,
    n_repeats=2,
    random_state=42
)

# Randomized Search
random_search = RandomizedSearchCV(
    estimator=model,
    param_distributions=param_dist,
    n_iter=40,   # puedes subir a 80 si tienes tiempo
    scoring='neg_root_mean_squared_error',
    cv=cv,
    verbose=1,
    random_state=42,
    n_jobs=-1
)

# Entrenamiento
random_search.fit(X_train, y_train)

# Resultados
print("Mejores hiperparámetros:")
print(random_search.best_params_)

print("\nMejor RMSE (CV):")
print(-random_search.best_score_)
````

````python
# ======================================== #
# MODELO FINAL CON MEJORES HIPERPARÁMETROS #
# ======================================== #

best_model = random_search.best_estimator_

y_pred = best_model.predict(X_test)

mae  = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2   = r2_score(y_test, y_pred)

print("------------------")
print("Resultados en TEST")
print("------------------")
print(f"MAE : {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R2  : {r2:.4f}")
````

<h1 align="center"><b>Underfitting y Overfitting</b></h1> 

</div>

---
<div style="text-align: justify; line-height: 1.5; font-size: 18px;">    

Analizar el **underfitting** y el **overfitting** permite entender si un modelo está aprendiendo de manera adecuada o si está fallando por exceso de simplicidad o complejidad. El objetivo es evaluar su capacidad de generalización, es decir, qué tan bien se desempeña en datos nuevos y no vistos. Este análisis guía el ajuste de hiperparámetros y la selección del modelo, buscando un equilibrio óptimo donde se capture la estructura real de los datos sin aprender el ruido.

</div>

---

<div style="display: flex; gap: 20px;">

<div style="flex: 1;">

## 🔵 Underfitting (subajuste)

**El modelo es demasiado simple**

- No captura la relación real de los datos  

- Falla incluso en entrenamiento  

**Ejemplo:** Intentar explicar todo con una línea recta cuando la realidad es curva.

</div>

<div style="flex: 1;">

## 🔴 Overfitting (sobreajuste)

**El modelo es demasiado complejo**

- Aprende muy bien los datos de entrenamiento  

- Pero falla en datos nuevos  

**Ejemplo:** Es como memorizar respuestas en vez de entender el tema.

</div>

</div>

---

````python
# ======================= #
# ANÁLISIS DE OVERFITTING #
# ======================= #

# TRAIN
y_train_pred = best_model.predict(X_train)

rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))

# CROSS-VALIDATION
cv_scores = cross_val_score(
    estimator = best_model,
    X = X_train,
    y = y_train,
    scoring = 'neg_root_mean_squared_error',
    cv = 5,
    n_jobs = -1
)

rmse_cv = -cv_scores.mean()
std_cv  = cv_scores.std()

# TEST
y_test_pred = best_model.predict(X_test)

rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))

# RESULTADOS
print("-----------------------------")
print("COMPARACIÓN DE ERRORES (RMSE)")
print("-----------------------------")
print(f"Train : {rmse_train:.4f}")
print(f"CV    : {rmse_cv:.4f} (+/- {std_cv:.4f})")
print(f"Test  : {rmse_test:.4f}")
````
